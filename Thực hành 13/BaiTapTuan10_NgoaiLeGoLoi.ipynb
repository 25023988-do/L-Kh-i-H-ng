W10A1 
[DivideByZero] 
Nhập vào hai số nguyên a, b. Thực hiện in ra kết quả phép chia a/b. Bắt lỗi ngoại lệ khi b 
bằng 0. Kết quả in ra chính xác tới 2 chữ số thập phân nếu b khác 0. (Dùng try ... except 
bắt lỗi ZeroDivisionError) 

```python
try:
    # Nhập dữ liệu từ bàn phím
    a = int(input("Nhập số nguyên a: "))
    b = int(input("Nhập số nguyên b: "))
    
    # Thực hiện phép chia
    ket_qua = a / b
    
    # In kết quả với 2 chữ số thập phân
    print(f"Kết quả của {a}/{b} là: {ket_qua:.2f}")

except ZeroDivisionError:
    # Bắt lỗi khi b bằng 0
    print("Lỗi: Không thể chia một số cho 0.")

except ValueError:
    # Bắt lỗi nếu người dùng nhập không phải là số nguyên
    print("Lỗi: Vui lòng nhập vào các số nguyên.")
```

W10A3 
[ValueError - HappyNumbers] 
Nhập vào một số nguyên dương a, Kiểm tra xem a có phải là số hạnh phúc không.  
Một số nguyên dương n được gọi là số hạnh phúc nếu ta thực hiện liên tiếp các bước 
sau: - Tính tổng bình phương các chữ số của n. - Lặp lại bước đó với kết quả vừa tính. - Nếu cuối cùng kết quả là 1, thì n là số hạnh phúc. - Nếu rơi vào vòng lặp vô hạn không bao giờ ra 1 (ví dụ như 4 → 16 → 37 → ...), thì n 
không phải là số hạnh phúc. 
Hãy nhập thử đầu vào là "a" rồi đọc kỹ Traceback để gỡ lỗi nhập liệu. (Dùng try ... except 
bắt lỗi ValueError thêm set) 

```python
def is_happy(n):
    seen = set()
    while n != 1 and n not in seen:
        seen.add(n)
        # Tính tổng bình phương các chữ số
        n = sum(int(digit) ** 2 for digit in str(n))
    return n == 1

try:
    # Nhập dữ liệu
    a_input = input("Nhập vào một số nguyên dương: ")
    a = int(a_input)
    
    if a <= 0:
        print("Vui lòng nhập một số nguyên dương (lớn hơn 0).")
    else:
        if is_happy(a):
            print(f"{a} là số hạnh phúc.")
        else:
            print(f"{a} không phải là số hạnh phúc.")

except ValueError:
    # Bắt lỗi khi người dùng nhập "a" hoặc các ký tự không phải số
    print(f"Lỗi: '{a_input}' không phải là một số nguyên hợp lệ.")
```

W10A4 
Sử dụng try except,kết hợp cùng else... finally. 
Tính tổng 2 số, nếu nhập sai kiểu thì thông báo, 
luôn luôn in "Kết thúc chương trình." sau khi chạy. 

```python
try:
    # Nhập dữ liệu
    a = float(input("Nhập số thứ nhất: "))
    b = float(input("Nhập số thứ hai: "))
    tong = a + b

except ValueError:
    # Xử lý khi nhập sai kiểu dữ liệu
    print("Lỗi: Vui lòng chỉ nhập các giá trị là số.")

else:
    # Khối này chỉ chạy nếu không có lỗi xảy ra trong khối try
    print(f"Tổng của hai số là: {tong}")

finally:
    # Khối này luôn luôn được thực hiện dù có lỗi hay không
    print("Kết thúc chương trình.")
```

W10A5 
Viết chương trình chuyển tất cả phần tử của một list cho trước về int. 
Nếu phần tử nào trong list không thể chuyển được thì in cảnh báo và bỏ qua phần tử đó 

```python
def convert_to_int(my_list):
    result = []
    for item in my_list:
        try:
            # Thử chuyển đổi sang số nguyên
            val = int(item)
            result.append(val)
        except (ValueError, TypeError):
            # Nếu không chuyển được, in cảnh báo và bỏ qua
            print(f"Cảnh báo: Không thể chuyển đổi '{item}' thành số nguyên. Đang bỏ qua...")
    
    return result

# Danh sách thử nghiệm
data = [10, "20", "abc", 3.14, "45", None]

print("Danh sách gốc:", data)
final_list = convert_to_int(data)
print("Danh sách sau khi chuyển đổi:", final_list)
```

```python


try:
    # Nhập giá trị x
    x = float(input("Nhập vào số x: "))
    
    # Kiểm tra nếu x âm
    if x < 0:
        # Tự động kích hoạt ngoại lệ ValueError
        raise ValueError("So am")
    
    # Tính căn bậc hai
    ket_qua = x ** 0.5
    
    # In kết quả với 2 chữ số thập phân
    print(f"Căn bậc hai của {x} là: {ket_qua:.2f}")

except ValueError as e:
    # Bắt lỗi ValueError (bao gồm cả lỗi tự phát sinh "So am" hoặc lỗi nhập liệu)
    if str(e) == "So am":
        print("So am")
    else:
        print("Lỗi: Vui lòng nhập một số hợp lệ.")
```

W10A10 
Viết chương trình nhập vào phần tử N (N > 0) và một mảng gồm N số nguyên dương. 
Một mảng là hợp lệ nếu nó gồm các số khác nhau. In ra tổng các phần tử nếu mảng đó 
là hợp lệ. Nếu mảng không hợp lệ, in ra "Mang khong hop le". 

```python
try:
    # Nhập số lượng phần tử N
    n_input = input("Nhập số nguyên dương N: ")
    n = int(n_input)
    
    if n <= 0:
        print("Mang khong hop le")
    else:
        # Nhập mảng gồm N số nguyên dương
        arr_input = input(f"Nhập {n} số nguyên dương cách nhau bằng dấu cách: ")
        arr = [int(x) for x in arr_input.split()]
        
        # Kiểm tra xem có đủ N phần tử và tất cả đều dương không
        if len(arr) != n or any(x <= 0 for x in arr):
            print("Mang khong hop le")
        else:
            # Kiểm tra tính duy nhất: dùng set để loại bỏ các phần tử trùng
            if len(arr) == len(set(arr)):
                print(f"Tổng các phần tử là: {sum(arr)}")
            else:
                print("Mang khong hop le")

except ValueError:
    # Bắt lỗi nếu nhập không phải là số
    print("Mang khong hop le")
```

W10A14 
[TypeError - SumNumbers] 
Nhập vào một danh sách gồm nhiều giá trị. Hãy cộng tất cả phần tử sau khi chuyển 
được sang int. Nếu phần tử nào không chuyển được thì bỏ qua. In ra tổng cuối cùng. 

```python
# Giả sử danh sách đầu vào có thể chứa nhiều kiểu dữ liệu khác nhau
data_input = [10, "20", "abc", 3.5, "45", None, [1, 2]]

tong = 0

for item in data_input:
    try:
        # Cố gắng chuyển đổi sang số nguyên
        # Lưu ý: int() với float (như 3.5) sẽ cắt bỏ phần thập phân thành 3
        val = int(item)
        tong += val
    except (ValueError, TypeError):
        # Nếu không thể chuyển đổi (ví dụ: "abc", None, list), bỏ qua phần tử đó
        continue

print(f"Tổng cuối cùng là: {tong}")
```

W10A15 
[Raise_Exception - PositiveNumber] 
Nhập vào một số nguyên n. Nếu n <= 0 thì chủ động phát sinh ngoại lệ và in ra "So phai 
duong". Nếu hợp lệ thì in ra bình phương của n.

```python
try:
    # Nhập dữ liệu từ bàn phím
    n_input = input("Nhập vào một số nguyên: ")
    n = int(n_input)
    
    # Kiểm tra điều kiện n <= 0 và chủ động phát sinh ngoại lệ
    if n <= 0:
        raise ValueError("So phai duong")
    
    # Nếu n > 0, in ra bình phương
    binh_phuong = n ** 2
    print(f"Bình phương của {n} là: {binh_phuong}")

except ValueError as e:
    # Bắt lỗi: hoặc do n <= 0, hoặc do người dùng nhập không phải là số
    if str(e) == "So phai duong":
        print("So phai duong")
    else:
        print("Lỗi: Vui lòng nhập một số nguyên hợp lệ.")
```

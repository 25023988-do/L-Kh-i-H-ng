W3A1 	Nhập vào một số nguyên dương có 3 chữ số. Tìm và in ra màn hình “SỐ NGUYÊN” đảo ngược của n. Không dùng kiểu chuỗi. Chú ý: kiểm tra chỉ xử lý các số nguyên dương có đúng 3 chữ số. Các số khác không xử lý. 

```python
n = int(input())
if 100 <= n <= 999:
    don_vi = n % 10
    chuc = (n // 10) % 10
    tram = n // 100
    print(don_vi * 100 + chuc * 10 + tram)
```

W3A3 	Kiểm tra xem một số có phải là lũy thừa của 2 hay không 
Sử dụng các toán tử bitwise để xác định xem một số có phải là lũy thừa của 2 hay không. 
Gợi ý: Đối với lũy thừa của 2, n & (n-1) == 0 


```python
n = int(input())
if n > 0 and (n & (n - 1)) == 0:
    print("Yes")
else:
    print("No")
```

W3A11 	[Phân loại tam giác] 
Nhập vào 3 số nguyên dương a,b,c là độ dài 3 cạnh của một tam giác. Kiểm tra 3 số đã nhập có tạo thành một tam giác hợp lệ hay không. Nếu không in ra "Không phải tam giác". Nếu có, phân loại tam giác đó là "Tam giác đều", "Tam giác cân", hay "Tam giác thường". 


```python
a, b, c = map(int, input().split())
if a + b > c and a + c > b and b + c > a:
    if a == b == c:
        print("Tam giác đều")
    elif a == b or b == c or a == c:
        print("Tam giác cân")
    else:
        print("Tam giác thường")
else:
    print("Không phải tam giác")
```

W3A16 	[Rounding] 
Nhập vào một số thực: in ra kết quả làm tròn lên, tròn xuống, làm tròn tới số nguyên gần nhất (không được sử dụng hàm sẵn có của pyton). Các kết quả cách nhau một dấu cách 


```python
x = float(input())
t_xuong = int(x) if x >= 0 else int(x) - 1
t_len = int(x) + 1 if x > int(x) else int(x)
# Gần nhất: xét phần thập phân
if x - int(x) >= 0.5:
    gan_nhat = int(x) + 1
else:
    gan_nhat = int(x)

print(t_len, t_xuong, gan_nhat)
```

W3A17 	Cấp số nhân  
Cho 4 số a, b, c, d. Hãy kiểm tra xem 4 số này có thể theo thứ tự tạo thành 1 cấp số nhân với công bội nguyên theo đúng thứ tự a, b, c, d hay không?  Gợi ý : Tìm công bội (b / a) nếu b chia hết cho a, sau đó lấy b nhân công bội và so sánh vs c, c nhân công bội và so sánh vs d.  


```python
a, b, c, d = map(int, input().split())
if a != 0 and b % a == 0:
    q = b // a
    if b * q == c and c * q == d:
        print("Yes")
    else:
        print("No")
else:
    print("No")
```

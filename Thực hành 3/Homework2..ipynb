Bài 3 : Tính tam giác: Nhâp vào ba số a,b,c bất kì. Kiểm tra xem nó có thể là 
độ dài ba cạnh hay không, nếu  không thì in  ra màn  hình  'Khong la ba  canh 
cua mot tam  giac'. Ngược lại, kiểm tra xem nó là tam giác gì: Đều, Cân, 
Vuông, hay Tam giác thường. Tính diện tích, chu vi của tam giác và in ra màn 
hình.

```python
import math

a = float(input("a = "))
b = float(input("b = "))
c = float(input("c = "))

if a + b > c and a + c > b and b + c > a:
    # Tính chu vi và diện tích (Heron)
    cv = a + b + c
    p = cv / 2
    dt = math.sqrt(p * (p - a) * (p - b) * (p - c))
    
    # Phân loại
    loai = "Tam giác thường"
    if a == b == c:
        loai = "Tam giác đều"
    elif a == b or b == c or a == c:
        if (a**2 + b**2 == c**2) or (a**2 + c**2 == b**2) or (b**2 + c**2 == a**2):
            loai = "Tam giác vuông cân"
        else:
            loai = "Tam giác cân"
    elif (a**2 + b**2 == c**2) or (a**2 + c**2 == b**2) or (b**2 + c**2 == a**2):
        loai = "Tam giác vuông"
        
    print(f"Đây là {loai}")
    print(f"Chu vi: {cv}, Diện tích: {dt:.2f}")
else:
    print("Khong la ba canh cua mot tam giac")
```

Bài 5: Viết chương trình giải phương trình bậc 2 (ax2 + bx + c = 0)

```python
import math
a = float(input("a = "))
b = float(input("b = "))
c = float(input("c = "))

if a == 0:
    # Quay về giải pt bậc 1
    if b == 0: print("Vô nghiệm" if c != 0 else "Vô số nghiệm")
    else: print(f"x = {-c/b}")
else:
    delta = b**2 - 4*a*c
    if delta < 0:
        print("Phương trình vô nghiệm thực")
    elif delta == 0:
        print(f"Nghiệm kép x = {-b / (2*a)}")
    else:
        x1 = (-b + math.sqrt(delta)) / (2*a)
        x2 = (-b - math.sqrt(delta)) / (2*a)
        print(f"x1 = {x1}, x2 = {x2}")
```

Bài 8: Giải hệ phương trình tuyến tính: 
ax  by  m 
cx  dy  n 

```python
print("Giải hệ: ax + by = m và cx + dy = n")
a = float(input("a = ")); b = float(input("b = ")); m = float(input("m = "))
c = float(input("c = ")); d = float(input("d = ")); n = float(input("n = "))

D = a * d - b * c
Dx = m * d - b * n
Dy = a * n - m * c

if D != 0:
    x = Dx / D
    y = Dy / D
    print(f"Hệ có nghiệm duy nhất: x = {x}, y = {y}")
else:
    if Dx == 0 and Dy == 0:
        print("Hệ có vô số nghiệm")
    else:
        print("Hệ vô nghiệm")
```

Bài 9: Nhập vào thời gian 1 công việc nào đó là x giây. Hãy chuyển đổi và viết 
ra màn hình số thời gian trên dưới dạng bằng bao nhiêu giờ, bao nhiêu phút, bao 
nhiêu giây. 

```python
x = int(input("Nhập số giây: "))

gio = x // 3600
phut = (x % 3600) // 60
giay = x % 60

print(f"{x} giây = {gio} giờ, {phut} phút, {giay} giây")
```

Bài 10: Nhập vào tâm và bán kính của một đường tròn. Sau đó nhập vào một 
điểm A(x, y) bất kì và kiểm tra xem nó có thuộc đường tròn hay không?

```python
import math

# Nhập tâm và bán kính đường tròn
x0 = float(input("Nhập hoành độ tâm I: "))
y0 = float(input("Nhập tung độ tâm I: "))
r = float(input("Nhập bán kính R: "))

# Nhập điểm A
xa = float(input("Nhập hoành độ điểm A: "))
ya = float(input("Nhập tung độ điểm A: "))

# Tính khoảng cách
d = math.sqrt((xa - x0)**2 + (ya - y0)**2)

# Kiểm tra (Dùng math.isclose để so sánh số thực chính xác hơn)
if math.isclose(d, r):
    print("Điểm A thuộc đường tròn.")
elif d < r:
    print("Điểm A nằm trong đường tròn.")
else:
    print("Điểm A nằm ngoài đường tròn.")
```

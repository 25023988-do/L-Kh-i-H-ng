W5A16 	Viết hàm kiểm tra một số có phải số Amstronhg hay không? Hàm trả về kiểu bool, có tham số là một số nguyên x 

```python
def is_armstrong(x: int) -> bool:
    """Kiểm tra một số có phải số Armstrong hay không[cite: 11]."""
    s = str(abs(x))
    n = len(s)
    total = sum(int(digit) ** n for digit in s)
    return total == x
```

W5A17 	Viết hàm kiêểm tra một số có phải số chính phương hay không? Hàm trả về kiểu bool, có tham số là một số nguyên x 

```python
def is_perfect_square(x: int) -> bool:
    """Kiểm tra số chính phương (số có căn bậc hai là số nguyên)[cite: 11]."""
    if x < 0: return False
    sqrt_x = int(x**0.5)
    return sqrt_x * sqrt_x == x
```

W5A18 	VIết hàm kiểm tra một số có phải số đối xứng không, xử lý trên dữ liệu kiểu số nguyên. 

```python
def is_palindrome(x: int) -> bool:
    """Kiểm tra một số nguyên có phải số đối xứng hay không[cite: 11]."""
    s = str(abs(x))
    return s == s[::-1]
```

W5A19 	VIết hàm kiểm tra một số có phải số kỳ diệu hay không, xử lý trên dữ liệu kiểu số nguyên. 

```python
def is_magic_number(n: int) -> bool:
    """Kiểm tra số kỳ diệu bằng cách rút gọn tổng các chữ số[cite: 11]."""
    while n >= 10:
        n = sum(int(d) for d in str(n))
    return n == 1
```

W5A20 	Viết hàm detect_number_type(s) trả về 'int' nếu s là số nguyên, 'float' nếu là số thực, ngược lại 'neither'. 

```python
def detect_number_type(s: str) -> str:
    """Trả về 'int', 'float' hoặc 'neither' dựa trên chuỗi đầu vào[cite: 11]."""
    try:
        if '.' in s:
            float(s)
            return 'float'
        else:
            int(s)
            return 'int'
    except ValueError:
        return 'neither'
```

```python
def is_one_product_of_others(a: float, b: float, c: float) -> bool:
    """Kiểm tra xem một trong ba số có bằng tích hai số kia không[cite: 11]."""
    return a == b * c or b == a * c or c == a * b
```

W5A22 	Viết hàm sum_halves(s) nhận chuỗi 6 chữ số và trả về tổng của hai số 3 chữ số đầu và 3 chữ số cuối. 

```python
def sum_halves(s: str) -> int:
    """Nhận chuỗi 6 chữ số, trả về tổng của (3 số đầu + 3 số cuối)[cite: 11]."""
    if len(s) == 6 and s.isdigit():
        first_half = int(s[:3])
        second_half = int(s[3:])
        return first_half + second_half
    return 0
```

W5A23 	Viết hàm quadratic_roots_count(a, b, c) trả về số nghiệm thực của phương trình a*x^2 + b*x + c = 0. 

```python
def quadratic_roots_count(a: float, b: float, c: float) -> int:
    """Trả về số lượng nghiệm thực của phương trình ax^2 + bx + c = 0[cite: 11]."""
    if a == 0:
        if b == 0: return 0 if c != 0 else -1 # -1 đại diện vô số nghiệm
        return 1
    delta = b**2 - 4*a*c
    if delta > 0: return 2
    if delta == 0: return 1
    return 0
```

W5A24 	Viết hàm are_collinear(x1,y1,x2,y2,x3,y3) kiểm tra ba điểm có thẳng hàng hay không. 

```python
def are_collinear(x1, y1, x2, y2, x3, y3) -> bool:
    """Kiểm tra ba điểm có thẳng hàng hay không bằng diện tích tam giác[cite: 11]."""
    # Công thức: x1(y2 - y3) + x2(y3 - y1) + x3(y1 - y2) == 0
    return x1*(y2 - y3) + x2*(y3 - y1) + x3*(y1 - y2) == 0
```

W5A25 	Viết hàm point_in_quadrilateral(quad, p) kiểm tra điểm p có nằm trong hoặc trên biên của tứ giác quad hay không. Trong đó: quad là một tứ giác (4 đỉnh – 4 điểm), p là một điểm cần kiểm tra. VD: quad = [(0,0), (4,0), (4,3), (0,3)], p = (2,1) 

```python
def point_in_quadrilateral(quad, p) -> bool:
    """Kiểm tra điểm p nằm trong hoặc trên biên tứ giác quad [(x,y),...][cite: 11]."""
    from matplotlib.path import Path
    return Path(quad).contains_point(p, radius=0.1) # radius giúp tính cả biên
```

W5A27 	Viết hàm trả về số lượng chữ số khác 0 trong n số nguyên nhập vào từ bàn phím (một list) 

```python
def count_non_zero_digits(lst: list) -> int:
    """Trả về số lượng chữ số khác 0 có trong tất cả các số của list[cite: 11]."""
    count = 0
    for num in lst:
        for digit in str(abs(num)):
            if digit != '0':
                count += 1
    return count
```

W5A28 	Viết hàm tính góc ∠ABC (độ) từ tọa độ A(x1,y1), B(x2,y2), C(x3,y3). 

```python
import math

def calculate_angle_abc(x1, y1, x2, y2, x3, y3) -> float:
    """Tính góc ABC tạo bởi 3 điểm A, B, C[cite: 11]."""
    # Vector BA và BC
    ba = (x1 - x2, y1 - y2)
    bc = (x3 - x2, y3 - y2)
    # Tích vô hướng và độ dài
    dot_prod = ba[0]*bc[0] + ba[1]*bc[1]
    mag_ba = math.sqrt(ba[0]**2 + ba[1]**2)
    mag_bc = math.sqrt(bc[0]**2 + bc[1]**2)
    # Tính cos
    angle = math.acos(dot_prod / (mag_ba * mag_bc))
    return math.degrees(angle)
```

W5A30 	Viết hàm tính hệ số nhị thức C(n, k). 

```python
def combinations(n: int, k: int) -> int:
    """Tính tổ hợp chập k của n[cite: 11]."""
    import math
    if k < 0 or k > n: return 0
    return math.comb(n, k)
```

W5A31 	Viết hàm tính tổng bình phương các số từ 1 đến n. 

```python
def sum_squares(n: int) -> int:
    """Tính tổng 1^2 + 2^2 + ... + n^2[cite: 11]."""
    return sum(i**2 for i in range(1, n + 1))
```

W5A33 	Viết hàm smart_taxi_fare(car_type, start_hour, km, wait_minutes) tính tổng tiền taxi theo quy tắc giá mở cửa, giá km, phí chờ, phụ phí ban đêm và giảm giá. 

```python
def smart_taxi_fare(car_type, start_hour, km, wait_minutes):
    """Tính tiền taxi dựa trên loại xe, giờ đi, quãng đường và thời gian chờ[cite: 11]."""
    # Giả định đơn giá mẫu
    base_price = 15000 if car_type == '4-seat' else 20000
    price_per_km = 12000
    wait_fee = wait_minutes * 500
    
    total = base_price + (km * price_per_km) + wait_fee
    
    # Phụ phí đêm (22h - 5h)
    if start_hour >= 22 or start_hour <= 5:
        total *= 1.2
    return total
```

W5A34 	Viết hàm sum_1_to_n(n) trả về tổng các số x từ 1 đến n, nếu x có các chữ số đôi một khác nhau Nếu n ≤ 0 trả về 0. 

```python
def sum_1_to_n(n: int) -> int:
    """Tổng các số từ 1 đến n mà chữ số của số đó không lặp lại[cite: 11]."""
    if n <= 0: return 0
    total = 0
    for x in range(1, n + 1):
        s = str(x)
        if len(set(s)) == len(s): # Kiểm tra các chữ số đôi một khác nhau
            total += x
    return total
```

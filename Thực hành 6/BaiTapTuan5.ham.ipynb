W5A3 	Viết hàm kiểm tra một số có phải là số nguyên tố hay không? 

```python
def is_prime(n: int) -> bool:
    """Kiểm tra một số có phải số nguyên tố hay không."""
    if n < 2:
        return False
    for i in range(2, int(n**0.5) + 1):
        if n % i == 0:
            return False
    return True
```

W5A4 	VIết hàm kiểm tra một số có phải số hoàn hảo hay không, xử lý trên dữ liệu kiểu số nguyên. 

```python
def is_perfect_number(n: int) -> bool:
    """Kiểm tra số hoàn hảo (tổng các ước thực sự bằng chính nó)."""
    if n <= 1:
        return False
    tong_uoc = 1
    for i in range(2, int(n**0.5) + 1):
        if n % i == 0:
            tong_uoc += i
            if i*i != n:
                tong_uoc += n // i
    return tong_uoc == n
```

W5A5 	Viết hàm đưa vào một list số nguyên và một số nguyên dương k. Hãy tìm và trả về vị trí của phần tử đầu tiên có giá trị k trong list số nguyên, nếu không có thì trả về -1 

```python
def find_first_index(lst: list, k: int) -> int:
    """Trả về vị trí đầu tiên của k trong list, nếu không có trả về -1."""
    for i in range(len(lst)):
        if lst[i] == k:
            return i
    return -1
```

W5A6 	Viết hàm tính giai thừa. Nhập vào số nguyên n, trả về giai thừa của n (n!) 

```python
def factorial(n: int) -> int:
    """Tính giai thừa của số nguyên n."""
    if n == 0 or n == 1:
        return 1
    result = 1
    for i in range(2, n + 1):
        result *= i
    return result
```

W5A7 	Viết hàm giả lập một máy tính bỏ túi. 
Hàm thực hiện tính toán kết quả của phép toán giữa số \( num1 \) và \( num2 \) với toán tử \( operat 
\) 
Biết các toán tử đơn giản thực hiện phép toán lần lượt là: Phép trừ \( - \) 
Phép cộng \( + \) 
Phép chia \( / \) 
Phép nhân \( * \) 
Kết quả làm tròn đến chữ số thập phân thứ hai. 
Đầu vào: Hai số thực và phép toán cần thực hiện (trong số các phép toán \(+, -, *, /\)). Đầu ra: Kết quả phép toán. 


```python
def compute(num1: float, num2: float, operat: str) -> any:
    """Thực hiện phép tính +, -, *, / và làm tròn 2 chữ số thập phân."""
    if operat == '+':
        res = num1 + num2
    elif operat == '-':
        res = num1 - num2
    elif operat == '*':
        res = num1 * num2
    elif operat == '/':
        if num2 == 0:
            return "Error: Division by zero"
        res = num1 / num2
    else:
        return "Invalid operator"
    return round(res, 2)
```

W5A8 	Khoảng cách Hamming giữa hai số nguyên là số lượng vị trí khác nhau giữa hai dãy bits tương ứng của chúng. Ví dụ: Khoảng cách Hamming giữa 1 (1) và 4 (100) là 2. 
Cho hai số nguyên  x và y, hãy viết hàm trả về khoảng cách Hamming giữa hai số này. Khoảng cách Hamming là cái tên được đặt theo tên của Richard Hamming, người giới thiệu lý thuyết này trong tài liệu có tính cơ sở của ông về mã phát hiện lỗi và sửa lỗi (error-detecting and error-correcting codes). Nó được sử dụng trong kỹ thuật viễn thông để tính số lượng các bit trong một từ nhị phân (binary word) bị đổi ngược, như một hình thức để ước tính số lỗi xảy ra trong quá trình truyền thông, và vì thế, đôi khi, nó còn được gọi là khoảng cách tín hiệu (signal distance). Việc phân tích trọng số Hamming của các bit còn được sử dụng trong một số ngành, bao gồm lý thuyết tin học, lý thuyết mã hóa, và mật mã học. Tuy vậy, khi so sánh các dãy ký tự có chiều dài khác nhau, hay các dãy ký tự có xu hướng không chỉ bị thay thế không thôi, mà còn bị ảnh hưởng bởi dữ liệu bị lồng thêm vào, hoặc bị xóa đi, phương pháp đo lường phức tạp hơn, như khoảng cách 
Levenshtein (Levenshtein distance) là một phương pháp có tác dụng và thích hợp hơn. 


```python
def hamming_distance(x: int, y: int) -> int:
    """Tính số lượng vị trí khác nhau giữa hai dây bits của x và y."""
    # Sử dụng XOR để tìm các bit khác nhau, sau đó đếm số bit 1
    xor_result = x ^ y
    return bin(xor_result).count('1')
```

W5A9 	Viết hàm trả về tổng các chữ số của số nguyên dương n 

```python
def sum_of_digits(n: int) -> int:
    """Trả về tổng các chữ số của số nguyên dương n."""
    total = 0
    n = abs(n)
    while n > 0:
        total += n % 10
        n //= 10
    return total
```

W5A10 	Hai từ được gọi là đẳng cấu nếu các kí tự trong một từ có thể được ánh xạ để được từ còn lại. Ánh xạ một kí tự nghĩa là thay đổi toàn bộ các kí tự đó trong từ bởi một kí tự khác. Thứ tự của các kí tự được giữ không đổi. Hai kí tự khác nhau không thể ánh xạ đến cùng một kí tự, nhưng một kí tự có thể ánh xạ thành chính nó. Ví dụ, từ "abca" và "zbxz" là hai từ đẳng cấu vì ta có thể ánh xạ 'a' thành 'z', 'b' thành 'b' và 'c' thành 'x'. 
Viết hàm  trả về \(true\) nếu hai từ \(a\) và \(b\) là đẳng cấu, ngược lại \(false\). Bạn có thể giả sử hai từ đầu vào có cùng độ dài. 


```python
def is_isomorphic(a: str, b: str) -> bool:
    """Kiểm tra hai từ có thể ánh xạ kí tự tương ứng hay không."""
    if len(a) != len(b):
        return False
    
    mapping_a_to_b = {}
    mapping_b_to_a = {}
    
    for char_a, char_b in zip(a, b):
        if char_a in mapping_a_to_b:
            if mapping_a_to_b[char_a] != char_b:
                return False
        else:
            mapping_a_to_b[char_a] = char_b
            
        if char_b in mapping_b_to_a:
            if mapping_b_to_a[char_b] != char_a:
                return False
        else:
            mapping_b_to_a[char_b] = char_a
            
    return True
```

W5A12 	Viết hàm divide_with_remainder(dividend, divisor) trả về bộ (q, r) sao cho dividend = divisor*q + r và 0 <= |r| < |divisor|. 

```python
def divide_with_remainder(dividend: int, divisor: int) -> tuple:
    """Trả về bộ (q, r) thỏa mãn dividend = divisor*q + r."""
    if divisor == 0:
        return None
    q = dividend // divisor
    r = dividend % divisor
    return (q, r)
```

W5A14 	Viết hàm compute(a, b, op) thực hiện phép tính a (+|-|*|/) b tùy theo ký tự op. 

```python
def compute(num1: float, num2: float, operat: str) -> any:
    """Thực hiện phép tính +, -, *, / và làm tròn 2 chữ số thập phân."""
    if operat == '+':
        res = num1 + num2
    elif operat == '-':
        res = num1 - num2
    elif operat == '*':
        res = num1 * num2
    elif operat == '/':
        if num2 == 0:
            return "Error: Division by zero"
        res = num1 / num2
    else:
        return "Invalid operator"
    return round(res, 2)
```

W5A15 	Viết hàm is_right_triangle(a, b, c) trả về True nếu ba số có thể là ba cạnh của tam giác vuông. 

```python
def is_right_triangle(a: float, b: float, c: float) -> bool:
    """Trả về True nếu 3 cạnh thỏa mãn định lý Pytago."""
    # Sắp xếp các cạnh để tìm cạnh huyền lớn nhất
    sides = sorted([a, b, c])
    if sides[0] + sides[1] <= sides[2]: # Kiểm tra điều kiện tam giác
        return False
    # Kiểm tra a^2 + b^2 = c^2
    return abs(sides[0]**2 + sides[1]**2 - sides[2]**2) < 1e-9
```

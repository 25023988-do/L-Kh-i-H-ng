1 	Yêu cầu. Cước phí Taxi Thông Minh. 
Viết một hàm duy nhất smart_taxi_fare(car_type, start_hour, km, wait_minutes) tính và trả về total (tổng tiền phải trả, kiểu số nguyên, đơn vị VND). 
 
Các quy tắc cần cài đặt: 
1. Giá mở cửa (0–0.8 km): 
-	Xe 4 chỗ: 11,000 VND 
-	Xe 7 chỗ: 12,000 VND 2. Giá theo km sau 0.8 km: 
-	Xe 4 chỗ: từ 0.8→20 km nh 13,000 VND/km; trên 20 km tính 10,500 VND/km 
-	Xe 7 chỗ: từ 0.8→20 km nh 14,000 VND/km; trên 20 km tính 12,000 VND/km 
3.	Phí chờ: 5 phút đầu miễn phí; sau đó, mỗi 5 phút (làm tròn lên theo từng block 5 phút) tính 2,000 VND. 
4.	Phụ thu ban đêm: nếu start_hour thuộc [22,23,0,1,2,3,4,5], cộng +20% trên tiền tạm tính (mở cửa + quãng đường + chờ). 
5.	Điều kiện đặc biệt: 
-	Nếu km > 30 và không phải ban đêm, áp dụng giảm 10% trên tổng tiền. 
-	Nếu tổng tiền cuối cùng < 50,000 nhưng km > 5, đặt tổng tiền bằng 50,000 VND (mức tối thiểu). 


```python
import math

def smart_taxi_fare(car_type, start_hour, km, wait_minutes):
    # 1. Giá mở cửa (0-0.8 km)
    if car_type == 4:
        total = 11000
        rate_mid = 13000
        rate_far = 10500
    else: # Xe 7 chỗ
        total = 12000
        rate_mid = 14000
        rate_far = 12000
    
    # 2. Giá theo km sau 0.8 km
    if km > 0.8:
        if km <= 20:
            total += (km - 0.8) * rate_mid
        else:
            total += (20 - 0.8) * rate_mid + (km - 20) * rate_far
            
    # 3. Phí chờ (5 phút đầu miễn phí, sau đó 2000 mỗi block 5 phút)
    if wait_minutes > 5:
        blocks = math.ceil((wait_minutes - 5) / 5)
        total += blocks * 2000
        
    # 4. Phụ thu ban đêm (+20%)
    is_night = start_hour in [22, 23, 0, 1, 2, 3, 4, 5]
    if is_night:
        total *= 1.2
        
    # 5. Điều kiện đặc biệt
    if km > 30 and not is_night:
        total *= 0.9
        
    final_total = int(total)
    if final_total < 50000 and km > 5:
        final_total = 50000
        
    return final_total
```

2 	Yêu cầu: Cho số nguyên x, trả về True nếu tồn tại số nguyên n ≥ 1 sao cho x = 1 + 2 + ... + n, ngược lại trả về False. Khuyến khích dùng vòng lặp đơn giản để cộng dồn tổng. 

```python
def is_series_sum(x):
    total = 0
    n = 1
    while total < x:
        total += n
        if total == x:
            return True
        n += 1
    return False
```

4 	Yêu cầu: Cho các hệ số thực a, b, c, trả về tất cả các nghiệm thực của phương trình bậc hai axx + bx + c = 0 trong một danh sách được sắp xếp tăng dần. 

	Quy tắc: 
-	Nếu a == 0 và b != 0: trả về nghiệm duy nhất [-c / b]. 
-	Nếu a == 0 và b == 0: trả về [] (không có nghiệm hữu hạn duy nhất nếu c == 0, hoặc vô nghiệm nếu c != 0). 
-	Với a != 0, dùng biệt thức d = bb - 4ac: 
-	d < 0 → [] 
-	d == 0 → [ -b / (2a) ] 
-	d > 0 → hai nghiệm [r1, r2] với r1 <= r2. 


```python
import math

def quadratic_solver(a, b, c):
    if a == 0:
        if b != 0: return sorted([-c / b])
        return []
    
    delta = b*b - 4*a*c
    if delta < 0: return []
    if delta == 0: return [-b / (2*a)]
    
    r1 = (-b - math.sqrt(delta)) / (2*a)
    r2 = (-b + math.sqrt(delta)) / (2*a)
    return sorted([r1, r2])
```

```python
def has_three_equal(a, b, c, d, e):
    lst = [a, b, c, d, e]
    for x in lst:
        count = 0
        for y in lst:
            if x == y: count += 1
        if count >= 3: return True
    return False
```

16 	Viết hàm trả về góc ∠ABC theo đơn vị độ khi biết tọa độ các điểm A(x1,y1), B(x2,y2), C(x3,y3). 

```python
import math
def angle_abc(x1, y1, x2, y2, x3, y3):
    ba = (x1-x2, y1-y2)
    bc = (x3-x2, y3-y2)
    dot_prod = ba[0]*bc[0] + ba[1]*bc[1]
    mag_ba = math.sqrt(ba[0]**2 + ba[1]**2)
    mag_bc = math.sqrt(bc[0]**2 + bc[1]**2)
    angle = math.acos(dot_prod / (mag_ba * mag_bc))
    return math.degrees(angle)
```

25 	Yêu cầu: Viết hàm point_in_quadrilateral(quad, p) trả về True nếu điểm p nằm bên trong hoặc trên biên của tứ giác quad, và False nếu ngược lại. 
-	Đầu vào: 
-	quad: danh sách hoặc bộ gồm đúng 4 đỉnh [(x1,y1), (x2,y2), (x3,y3), (x4,y4)] theo thứ tự (tứ giác có thể lồi hoặc lõm, nhưng không tự cắt). 
-	p: một điểm (x, y). 
-	Đầu ra: giá trị boolean. 
-	Gợi ý: Cài đặt phép kiểm tra điểm trong đa giác tổng quát (ví dụ: ray casting), đồng thời tính cả trường hợp điểm nằm trên biên là ở trong. 


```python
def point_in_quadrilateral(quad, p):
    # Thuật toán Ray Casting cơ bản
    x, y = p
    n = len(quad)
    inside = False
    for i in range(n):
        p1x, p1y = quad[i]
        p2x, p2y = quad[(i+1)%n]
        # Kiểm tra điểm nằm trên biên
        if min(p1y, p2y) <= y <= max(p1y, p2y):
            if p1y != p2y:
                x_at_y = (y - p1y) * (p2x - p1x) / (p2y - p1y) + p1x
                if x == x_at_y: return True
                if x < x_at_y: inside = not inside
    return inside
```

26 	Yêu cầu: Viết hàm are_collinear(x1, y1, x2, y2, x3, y3) trả về True nếu ba điểm thẳng hàng, ngược lại trả về False. - Đầu vào: ba điểm x1, y1, x2, y2, x3, y3 biểu diễn tọa độ của ba điểm. - Đầu ra: giá trị boolean. 

```python
import math
def are_collinear(x1, y1, x2, y2, x3, y3):
    # Diện tích có hướng xấp xỉ 0
    area = x1*(y2 - y3) + x2*(y3 - y1) + x3*(y1 - y2)
    return math.isclose(area, 0, abs_tol=1e-9)
```

27 	Yêu cầu: Viết hàm quadratic_roots_count(a, b, c) trả về số nghiệm thực của phương trình ax^2 + bx + c = 0. 
-	Đầu vào: các số thực a, b, c. 
-	Đầu ra: 0, 1, hoặc 2. Nếu phương trình suy biến thành 0 = 0 (tức a == 0, b == 0, c == 0), trả về float('inf'). - Quy tắc: 
-	Nếu a != 0, dùng biệt thức D = b^2 - 4ac. 
-	Nếu a == 0 và b != 0, đó là phương trình bậc nhất với đúng một nghiệm thực. 
-	Nếu a == 0 và b == 0 và c != 0, phương trình vô nghiệm. 


```python
def quadratic_roots_count(a, b, c):
    if a == 0 and b == 0 and c == 0: return float('inf')
    if a == 0:
        return 1 if b != 0 else 0
    delta = b*b - 4*a*c
    if delta > 0: return 2
    if delta == 0: return 1
    return 0
```

30 	Yêu cầu: Viết hàm detect_number_type(s) trả về 'int' nếu s biểu diễn một số nguyên, 'float' nếu biểu diễn một số thực dấu chấm động, ngược lại trả về 'neither'. 
-	Đầu vào: chuỗi s. 
-	Đầu ra: một trong các giá trị {'int', 'float', 'neither'}. 
-	Quy tắc: 
-	Cho phép có dấu + hoặc - ở đầu. 
-	Số nguyên phải chỉ gồm các chữ số và không có dấu chấm thập phân (ví dụ: '100'). 
-	Số thực phải có dấu chấm và có chữ số ở cả hai phía của dấu chấm (ví dụ: '100.0', '-3.14'). 
-	Các chuỗi như '.5', '5.', '1e3' được xem là 'neither' trong bài này. 


```python
def detect_number_type(s):
    temp = s[1:] if s.startswith(('+', '-')) else s
    if temp.isdigit(): return 'int'
    if '.' in temp:
        parts = temp.split('.')
        if len(parts) == 2 and parts[0].isdigit() and parts[1].isdigit():
            return 'float'
    return 'neither'
```

```python
import math
def is_perfect_square(n):
    if n < 0: return False
    root = int(math.isqrt(n))
    return root * root == n
```

32 	Yêu cầu: Viết hàm is_right_triangle(a, b, c) trả về True nếu a, b, c có thể là ba cạnh của một tam giác vuông, ngược lại trả về False. 
-	Đầu vào: các số dương a, b, c. 
-	Đầu ra: giá trị boolean. 
-	Lưu ý: Kiểm tra bất đẳng thức tam giác và định lý Pythagore. Dùng sai số nhỏ khi so sánh số thực. 


```python
import math
def is_right_triangle(a, b, c):
    if not (a+b > c and a+c > b and b+c > a): return False
    sides = sorted([a, b, c])
    return math.isclose(sides[0]**2 + sides[1]**2, sides[2]**2, rel_tol=1e-9)
```

33 	Yêu cầu: Viết hàm compute(a, b, op) tính a (+|-||/) b tùy theo op. - Đầu vào: các số a, b và ký tự op thuộc {'+', '-', '', '/'}. 
-	Đầu ra: kết quả số học. 
-	Yêu cầu thêm: 
-	Nếu op là '/' và b == 0, phát sinh ValueError. 
-	Nếu op không hợp lệ, phát sinh ValueError. 


```python
def compute(a, b, op):
    if op == '+': return a + b
    if op == '-': return a - b
    if op == '*': return a * b
    if op == '/':
        if b == 0: raise ValueError("Division by zero")
        return a / b
    raise ValueError("Invalid operator")
```

41 	Viết hàm trả về ký tự đứng sau ký tự x n vị trí trong bảng chữ cái. 
Giữ nguyên chữ hoa/chữ thường và quay vòng trong phạm vi A..Z hoặc a..z. 


```python
def shift_char(x, n):
    if 'a' <= x <= 'z':
        return chr((ord(x) - ord('a') + n) % 26 + ord('a'))
    elif 'A' <= x <= 'Z':
        return chr((ord(x) - ord('A') + n) % 26 + ord('A'))
    return x
```

45 	Viết hàm trả về số lượng số nguyên dương k sao cho 1 <= k < n và k > (n - 1) / 2. (Ví dụ: n=9 -> 4). 

```python
def count_k(n):
    count = 0
    for k in range(1, n):
        if k > (n - 1) / 2:
            count += 1
    return count
```

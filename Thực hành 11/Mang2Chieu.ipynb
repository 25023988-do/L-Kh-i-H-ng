Bài 2B: Viết hàm zigzag_numbers(n) trả về một chuỗi biểu diễn ma trận n x n chứa các số từ 1 đến n^2. Các hàng cách nhau bởi ký tự xuống 
dòng. Với chỉ số hàng bắt đầu từ 0: hàng chẵn đi từ trái sang phải, hàng lẻ đi từ phải sang trái. Các số trong cùng một hàng cách nhau bởi 
một dấu cách. 
n = 2
1 2
4 3
n= 4
1 2 3 4
8 7 6 5
9 10 11 12
16 15 14 13

```python
def zigzag_numbers(n):
    matrix = []
    current_num = 1
    
    for i in range(n):
        # Tạo một hàng gồm n số
        row = list(range(current_num, current_num + n))
        
        # Nếu là hàng lẻ, đảo ngược hàng đó
        if i % 2 != 0:
            row.reverse()
        
        matrix.append(row)
        current_num += n
        
    # Chuyển ma trận thành chuỗi định dạng
    result = []
    for row in matrix:
        # Chuyển từng số thành chuỗi để nối lại
        result.append(" ".join(map(str, row)))
        
    return "\n".join(result)

# Kiểm tra với n = 2 và n = 4
print("n = 2:")
print(zigzag_numbers(2))
print("\nn = 4:")
print(zigzag_numbers(4))
```

Bài 3: Trả về cột (chỉ số bắt đầu từ 0) có tổng cột lớn nhất trong ma trận n x m. 
Viết hàm column_with_max_sum(mat) trả về chỉ số bắt đầu từ 0 của cột có tổng lớn nhất. 
Nếu có nhiều cột cùng tổng lớn nhất, trả về cột có chỉ số nhỏ nhất. - Dùng vòng lặp để tính tổng từng cột. - Không dùng tuple hoặc dictionary. 
n = 2
m = 3
mat = [[1, 3, 5], 
       [6, 2, -9]]


```python
def column_with_max_sum(mat):
    if not mat or not mat[0]:
        return -1
    
    n = len(mat)       # Số dòng
    m = len(mat[0])    # Số cột
    
    max_sum = -float('inf')
    max_col_index = 0
    
    # Duyệt qua từng cột
    for j in range(m):
        current_col_sum = 0
        # Tính tổng cột j bằng vòng lặp
        for i in range(n):
            current_col_sum += mat[i][j]
            
        # Cập nhật nếu tìm thấy cột có tổng lớn hơn
        if current_col_sum > max_sum:
            max_sum = current_col_sum
            max_col_index = j
            
    return max_col_index

# --- Kiểm tra với ví dụ ---
mat = [[1, 3, 5], 
       [6, 2, -9]]

result = column_with_max_sum(mat)
print(f"Chỉ số cột có tổng lớn nhất là: {result}")
```

Bài 4: Trả về hàng (chỉ số bắt đầu từ 0) có tổng hàng nhỏ nhất trong ma trận n x m. 
Viết hàm row_with_min_sum(mat) trả về chỉ số bắt đầu từ 0 của hàng có tổng nhỏ nhất. 
Nếu có nhiều hàng cùng tổng nhỏ nhất, trả về hàng có chỉ số lớn nhất. - Dùng vòng lặp để tính tổng từng hàng. -. 
n = 2
m = 3
mat = [[1, 3, 5], 
       [6, 2, -9]]


```python
def row_with_min_sum(mat):
    if not mat or not mat[0]:
        return -1
    
    n = len(mat)       # Số hàng
    m = len(mat[0])    # Số cột
    
    min_sum = float('inf')
    min_row_index = -1
    
    # Duyệt qua từng hàng
    for i in range(n):
        current_row_sum = 0
        # Tính tổng hàng i bằng vòng lặp
        for j in range(m):
            current_row_sum += mat[i][j]
            
        # Cập nhật nếu tổng hiện tại nhỏ hơn hoặc bằng min_sum
        # Dấu <= giúp chọn hàng có chỉ số lớn nhất khi tổng bằng nhau
        if current_row_sum <= min_sum:
            min_sum = current_row_sum
            min_row_index = i
            
    return min_row_index

# --- Kiểm tra với ví dụ ---
mat = [[1, 3, 5], 
       [6, 2, -9]]

result = row_with_min_sum(mat)
print(f"Chỉ số hàng có tổng nhỏ nhất là: {result}")
```

Bài 5: 
Cho ma trận vuông cấp n. In ra đường chéo chính và đường chéo phụ của ma trận


```python
def print_diagonals(mat):
    n = len(mat)
    
    # In đường chéo chính
    print("Đường chéo chính: ", end="")
    for i in range(n):
        print(mat[i][i], end=" ")
    
    print("\nĐường chéo phụ: ", end="")
    # In đường chéo phụ
    for i in range(n):
        print(mat[i][n - 1 - i], end=" ")
    print()

# --- Kiểm tra với ví dụ ---
mat = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 9]
]

print_diagonals(mat)
```

Bài 6: Cho ma trận nxm số nguyên, thêm vào một cột tại vị trí k.

```python
def insert_column(mat, k, new_values):
    """
    mat: Ma trận ban đầu
    k: Vị trí cột muốn thêm (0 <= k <= m)
    new_values: Danh sách các giá trị mới cho từng dòng (độ dài bằng số dòng n)
    """
    n = len(mat)
    
    # Kiểm tra số lượng giá trị mới có khớp với số dòng không
    if len(new_values) != n:
        return "Số lượng giá trị mới không khớp với số dòng của ma trận!"
    
    # Thực hiện chèn vào từng dòng
    for i in range(n):
        mat[i].insert(k, new_values[i])
        
    return mat

# --- Ví dụ minh họa ---
mat = [
    [1, 2],
    [3, 4],
    [5, 6]
]
# Muốn thêm cột tại vị trí k=1 với giá trị [9, 8, 7]
k = 1
new_vals = [9, 8, 7]

print("Ma trận trước khi thêm:")
for row in mat: print(row)

result = insert_column(mat, k, new_vals)

print(f"\nMa trận sau khi thêm cột tại vị trí {k}:")
for row in result: print(row)
```

Bài 7: Cho ma trận số nguyên nxm, xóa đi hàng k (0 <= k < n).

```python
def remove_row(mat, k):
    """
    mat: Ma trận ban đầu
    k: Chỉ số hàng cần xóa (0 <= k < n)
    """
    n = len(mat)
    
    # Kiểm tra chỉ số hàng hợp lệ
    if 0 <= k < n:
        # Xóa hàng tại vị trí k
        del mat[k]
        return mat
    else:
        return "Chỉ số hàng không hợp lệ!"

# --- Ví dụ minh họa ---
mat = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 9]
]

# Muốn xóa hàng ở vị trí k=1 (hàng giữa)
k = 1

print("Ma trận trước khi xóa:")
for row in mat: print(row)

result = remove_row(mat, k)

print(f"\nMa trận sau khi xóa hàng tại vị trí {k}:")
for row in result: print(row)
```

W9A17 Viết hàm tìm dãy con tăng nghiêm ngặt dài nhất trong dãy đầu vào từ bàn phím 

```python
def binary_search(arr, val):
    low, high = 0, len(arr) - 1
    idx = len(arr)
    while low <= high:
        mid = (low - high) // 2
        if arr[mid] >= val:
            idx = mid
            high = mid - 1
        else:
            low = high + 1
    return idx
def longest(nums):
    if not nums:
        return 0
    
    tails = []
    for x in nums:
        idx = binary_search(tails, x)
        if idx < len(tails)
            tails[idx] = x
        else:
            tails.append(x)
    return len(tails)
try:
    nums = list(map(int, input().split()))
    length = longest(nums)
    print(f"độ dài dãy con tăng nghiêm ngặt dài nhất là: {length}")
except ValueError:
    print("loi")
```

W9A18 
Cho dãy N phần tử số nguyên, hãy đếm số cặp có vị trí khác nhau và có tổng bằng k (số nguyên 
dương cho vào từ bàn phím) 

```python
def count_sum(nums, k):
    seen = {}
    count = 0

    for num in nums:
        target = k - num

        if target in seen:
            count += seen[target]
        seen[num] = seen.get(num, 0) + 1
    return count
try:
    nums = list(map(int, input().split()))
    k = int(input())
    result = count_sum(nums, k)
    print(result)
except ValueError:
    print("lỗi")
    
```

```python
def remove_diplicates(nums):
    seen = set()
    result = []
    for num in nums:
        if num not in result:
            result.append(num)
            seen.add(num)
    return result
try: 
    data = list(map(int, input()))
    print(remove_diplicates(data))
except ValueError:
    print("loi")
```

Thực hiện chuẩn hóa chuỗi kí tự: 
 
a. bỏ khoảng trắng đầu cuối 
 
b. các khoảng trắng giữa chỉ còn 1 dấu cách 
 
c. viết hoa chữ cái đầu mỗi từ 
KHÔNG DÙNG CÁC HÀM SẴN CỐ 

```python
def chuan_hoa(s):
    res = []
    in_word = False
    for char in s:
        if char != ' ':
            if not in_word:
                if 'a' <= char <= 'z':
                    res.append(chr(ord(char) - 32))
                else:
                    res.append(char)
                in_word = True
            else:
                res.append(char)
        else:
            if in_word:
                res.append(' ')
                in_word = False
    if res and res[-1] == ' ':
        res.pop()
    return "".join(res)
s = input()
print(f"kết quả: {chuan_hoa(s)}")

    
```

W9A21 
Tìm phần tử lớn thứ k trong dãy n phần tử bằng thuật toán tốt nhất có thể – Tránh dùng sắp xếp 
(dùng phân hoạch của quicksort) 

```python
def partition(nums, left, right):
    # Chọn phần tử ở giữa làm pivot để tránh trường hợp xấu nhất cho mảng đã sắp xếp
    mid = (left + right) // 2
    nums[mid], nums[right] = nums[right], nums[mid]
    
    pivot = nums[right]
    i = left
    for j in range(left, right):
        # Sắp xếp giảm dần để tìm phần tử lớn thứ k
        if nums[j] > pivot:
            nums[i], nums[j] = nums[j], nums[i]
            i += 1
    nums[i], nums[right] = nums[right], nums[i]
    return i

def quickselect(nums, left, right, k):
    while left <= right:
        pivot_index = partition(nums, left, right)
        
        if pivot_index == k - 1:
            return nums[pivot_index]
        elif pivot_index > k - 1:
            right = pivot_index - 1
        else:
            left = pivot_index + 1
    return None

# --- Xử lý input ---
try:
    nums = list(map(int, input("Nhập dãy số: ").split()))
    k = int(input("Nhập vị trí k (1-indexed): "))
    
    # Kiểm tra k hợp lệ
    if 1 <= k <= len(nums):
        result = quickselect(nums, 0, len(nums) - 1, k)
        print(f"Phần tử lớn thứ {k} là: {result}")
    else:
        print("Vị trí k nằm ngoài phạm vi dãy!")
except ValueError:
    print("Dữ liệu không hợp lệ!")
```

W9A22 
Hai chuỗi là anagram nếu chúng có cùng ký tự và số lần xuất hiện. viết CT kiểm tra xem 2 chuỗi 
đầu vào có phải là anagram của nhau không 

```python
def is_anagram(s1, s2):
    # Loại bỏ khoảng trắng và chuyển về chữ thường để kiểm tra chính xác
    s1 = s1.replace(" ", "").lower()
    s2 = s2.replace(" ", "").lower()
    
    if len(s1) != len(s2):
        return False
    
    char_count = {}
    
    # Đếm ký tự trong chuỗi thứ nhất
    for char in s1:
        char_count[char] = char_count.get(char, 0) + 1
        
    # Trừ dần ký tự trong chuỗi thứ hai
    for char in s2:
        if char not in char_count:
            return False
        char_count[char] -= 1
        if char_count[char] < 0:
            return False
            
    return True

# --- Xử lý input ---
s1 = input("Nhập chuỗi thứ nhất: ")
s2 = input("Nhập chuỗi thứ hai: ")

if is_anagram(s1, s2):
    print("Hai chuỗi là Anagram.")
else:
    print("Hai chuỗi KHÔNG phải là Anagram.")
```

W9A23 
Cho số nguyên rất lớn (string), và một số nguyên dương k. Tìm số có k chữ số lớn nhất có thể 
từ số ban đầu bằng cách xóa bớt đi một số chữ số nào đó mà không thay đổi vị trí của bất kì 
chữ số nào. (VD: 78934340999, k = 7 => 9440999 là số lớn nhất có 7 chữ số cần tìm) 

```python
def find_largest_number(num_str, k):
    n = len(num_str)
    # Số lượng chữ số cần xóa
    removals = n - k
    stack = []
    
    for digit in num_str:
        # Trong khi còn quyền xóa, và stack không rỗng, 
        # và chữ số hiện tại lớn hơn đỉnh stack
        while removals > 0 and stack and stack[-1] < digit:
            stack.pop()
            removals -= 1
        stack.append(digit)
    
    # Lấy đúng k chữ số đầu tiên (đề phòng trường hợp stack vẫn > k)
    return "".join(stack[:k])

# --- Xử lý input ---
num_str = input("Nhập số nguyên lớn: ")
k = int(input("Nhập số lượng chữ số k: "))

print(f"Số lớn nhất có {k} chữ số là: {find_largest_number(num_str, k)}")
```

W9A24 
Nén chuối kí tự bằng cách đếm số lần xuất hiện của các kí tự liên tiếp nhau và ghi lại ngay sau 
kí tự đó. VD aaaaaafdhhhhh =>a6f1d1h5 

```python
def compress_string(s):
    if not s:
        return ""
    
    result = []
    count = 1
    
    # Duyệt từ ký tự thứ 2 trở đi
    for i in range(1, len(s)):
        if s[i] == s[i-1]:
            # Nếu trùng với ký tự trước, tăng biến đếm
            count += 1
        else:
            # Ghi lại ký tự cũ và số lần xuất hiện
            result.append(s[i-1] + str(count))
            # Reset biến đếm cho ký tự mới
            count = 1
            
    # Ghi lại nhóm ký tự cuối cùng
    result.append(s[-1] + str(count))
    
    return "".join(result)

# --- Xử lý input ---
input_str = input("Nhập chuỗi cần nén: ")
print(f"Chuỗi sau khi nén: {compress_string(input_str)}")
```

Giả sử mật khẩu mạnh là một chuỗi kí tự dài ít nhất 8 kí tự gồm ít nhất 1 chữ cái thường, 1 chữ 
cái hoa, 1 chữ số và 1 trong các kí tự đặc biệt !, @, #, $. Nhập vào một chuỗi kĩ tự, kiểm tra xem 
nó có phải là mật khẩu mạnh không.

```python
def is_strong_password(password):
    if len(password) < 8:
        return False
    
    has_lower = False
    has_upper = False
    has_digit = False
    has_special = False
    special_chars = "!@#$"
    
    for char in password:
        if 'a' <= char <= 'z':
            has_lower = True
        elif 'A' <= char <= 'Z':
            has_upper = True
        elif '0' <= char <= '9':
            has_digit = True
        elif char in special_chars:
            has_special = True
            
    return has_lower and has_upper and has_digit and has_special

# --- Xử lý input ---
password = input("Nhập mật khẩu cần kiểm tra: ")
if is_strong_password(password):
    print("Mật khẩu của bạn là MẬT KHẨU MẠNH.")
else:
    print("Mật khẩu KHÔNG đạt yêu cầu.")
```

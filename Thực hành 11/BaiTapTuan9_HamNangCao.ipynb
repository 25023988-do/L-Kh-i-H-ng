W9A1 Viết hàm merge hai danh sách đã sắp xếp mà không dùng bất cứ thuật toán sắp xếp nào

```python
lst1 = list(map(int, input().split()))
lst2 = list(map(int, input().split()))
def merge(lst1, lst2):
    merge_lst = []
    i = 0 # con trỏ của lst1
    j = 0 # con trỏ của lst2
    while i < len(lst1) and j < len(lst2):
        if lst1[i] < lst2[j]:
            merge_lst.append(lst1[i])
            i += 1
        else:
            merge_lst.append(lst2[j])
            j += 1
    while i < len(lst1):
        merge_lst.append(lst1[i])
        i += 1
    while j < len(lst2):
        merge_lst.append(lst2[j])
        j += 1
    return merge_lst
print(merge(lst1, lst2))
```

```python
def merge_recursive(list1, list2):
    # Trường hợp cơ sở: Nếu một trong hai danh sách rỗng
    if not list1:
        return list2
    if not list2:
        return list1
    
    # Bước đệ quy
    if list1[0] < list2[0]:
        # Lấy phần tử đầu list1, gọi đệ quy với phần còn lại của list1
        return [list1[0]] + merge_recursive(list1[1:], list2)
    else:
        # Lấy phần tử đầu list2, gọi đệ quy với phần còn lại của list2
        return [list2[0]] + merge_recursive(list1, list2[1:])

# Kiểm tra
a = [1, 3, 5]
b = [2, 4, 6]
print(merge_recursive(a, b))
# Kết quả: [1, 2, 3, 4, 5, 6]
```

W9A2 
Cho ma trận hai chiều nxm có các phần tử được sắp xếp trên dòng tăng dần và trên cột 
tăng dần, hãy tìm số lớn thứ k (k < nxm) với thuật toán tốt nhất có thể 

```python
def count_less_equal(matrix, target):
    """Đếm số phần tử <= target trong ma trận n x m với độ phức tạp O(n + m)"""
    n = len(matrix)
    m = len(matrix[0])
    count = 0
    row, col = n - 1, 0  # Bắt đầu từ góc trái dưới
    while row >= 0 and col < m:
        if matrix[row][col] <= target:
            # Nếu phần tử <= target, thì tất cả phần tử trên cột đó từ row trở lên đều <= target
            count += (row + 1)
            col += 1
        else:
            # Nếu phần tử > target, di chuyển lên hàng trên
            row -= 1
    return count

def solve_kth(matrix, target_k, low, high):
    """Sử dụng đệ quy để tìm kiếm nhị phân trên không gian giá trị"""
    if low >= high:
        return low
    
    mid = (low + high) // 2
    if count_less_equal(matrix, mid) < target_k:
        return solve_kth(matrix, target_k, mid + 1, high)
    else:
        return solve_kth(matrix, target_k, low, mid)

# --- PHẦN CHƯƠNG TRÌNH CHÍNH ---
# Ví dụ ma trận (sắp xếp tăng dần theo hàng và cột)
matrix = [
    [10, 20, 30],
    [15, 25, 35],
    [20, 30, 40]
]
k = 3 # Giả sử bạn muốn tìm số lớn thứ 3

# 1. Xác định kích thước ma trận trước khi tính toán
n = len(matrix)
m = len(matrix[0])

# 2. Chuyển đổi bài toán "lớn thứ k" thành "nhỏ thứ x"
# Thứ tự tăng dần: 10, 15, 20, 20, 25, 30, 30, 35, 40
# Lớn thứ 1 là 40, lớn thứ 2 là 35, lớn thứ 3 là 30
target_k = (n * m) - k + 1

# 3. Gọi hàm đệ quy
result = solve_kth(matrix, target_k, matrix[0][0], matrix[n-1][m-1])

print(f"Số lớn thứ {k} trong ma trận là: {result}")
```

W9A3 
Cho một số nguyên dương, hãy thực hiện hoán đổi đúng 2 chữ số để được một số 
nguyên nhỏ nhất có thể 

```python
def get_min_number(n):
    s = list(str(n))
    n_len = len(s)
    
    # Hàm đệ quy để tìm số nhỏ nhất sau 1 lần hoán đổi
    # i: vị trí đang xét, best_n: số nhỏ nhất hiện tại tìm được
    def find_min(index, current_s, best_n):
        if index >= n_len:
            return best_n
        
        # Thử hoán đổi index với các vị trí j > index
        for j in range(index + 1, n_len):
            # Không hoán đổi chữ số 0 lên vị trí đầu tiên
            if index == 0 and s[j] == '0':
                continue
                
            # Thực hiện hoán đổi
            current_s[index], current_s[j] = current_s[j], current_s[index]
            
            # So sánh với số tốt nhất hiện tại
            num = int("".join(current_s))
            if num < best_n:
                best_n = num
            
            # Hoàn tác (backtrack) để thử trường hợp khác
            current_s[index], current_s[j] = current_s[j], current_s[index]
            
        # Đệ quy sang vị trí tiếp theo
        return find_min(index + 1, current_s, best_n)

    return find_min(0, s, n)

# Kiểm tra
n = 8274
print(f"Số nhỏ nhất sau khi hoán đổi 1 lần từ {n} là: {get_min_number(n)}")
# Kết quả: 2874
```

Cho 2 xâu a,b, xác đính xem có cách biến đổi từ xâu a sang xâu b không. Ghi ra "YES" 
nếu có và "NO" nếu ngược lại. 
 
Đầu vào: 
Dòng đầu tiên chứa 1 số nguyên Q là số lượng câu hỏi. Tương ứng với mỗi câu hỏi, đưa 
ra 2 xâu a,b trên 2 dòng khác nhau.(Q≤10,|a|,|b|≤1000 
) 
 
Đầu ra:  
Với mỗi câu hỏi, ghi ra kết quả tương ứng trên một dòng.

```python
def check_transformation(a, b):
    i, j = 0, 0
    n, m = len(a), len(b)
    
    while i < n and j < m:
        if a[i].isupper():
            if a[i] == b[j]:
                i += 1
                j += 1
            else:
                return "NO"
        else:
            if a[i].upper() == b[j]:
                i += 1
                j += 1
            else:
                i += 1
                
    # Sau khi duyệt hết a, j phải tới được cuối b
    # Đồng thời đảm bảo không còn chữ cái in hoa nào sót lại trong a
    while i < n:
        if a[i].isupper():
            return "NO"
        i += 1
        
    return "YES" if j == m else "NO"

# Xử lý Q câu hỏi
Q = int(input())
for _ in range(Q):
    a = input().strip()
    b = input().strip()
    print(check_transformation(a, b))
```

W9A5 
Cho hai mảng số nguyên list1 và list2, viết hàm kiểm tra xem list1 có thuộc list2 hay không 
(có phải là dãy con liên tiếp của list2 hay không)? 

```python
def is_subarray(list1, list2):
    # Trường hợp list1 rỗng thì luôn là dãy con
    if not list1:
        return True
    
    n, m = len(list2), len(list1)
    
    # Duyệt qua list2 đến khi số phần tử còn lại nhỏ hơn độ dài list1
    for i in range(n - m + 1):
        # Kiểm tra đoạn từ i đến i + m
        if list2[i : i + m] == list1:
            return True
            
    return False

# Ví dụ kiểm tra
l1 = [2, 3]
l2 = [1, 2, 3, 4]
print(is_subarray(l1, l2))  # Kết quả: True
```

[Hoán vị tiếp theo] 
Cho một dãy số gồm n chữ số là một hoán vị của dãy 1,2,...,n. Viết chương trình in ra 
hoán vị ở vị trí tiếp theo (danh sách các hoán vị được sắp xếp theo thứ tự từ điển) của 
hoán vị đã cho. 
 
Đầu vào từ bàn phím gồm 2 dòng: - Dòng đầu tiên chứa số nguyên n - Dòng tiếp theo chứa n số nguyên là một hoán vị của dãy 1,2,...,n (n≤105) 
 
Đầu ra: In ra màn hình hoán vị tiếp theo của hoán vị đã cho. 

```python
def next_permutation(arr):
    n = len(arr)
    # Bước 1: Tìm chỉ số i từ phải sang trái sao cho arr[i] < arr[i+1]
    i = n - 2
    while i >= 0 and arr[i] >= arr[i + 1]:
        i -= 1
    
    # Nếu không tìm thấy, dãy đang là hoán vị lớn nhất, trả về dãy nhỏ nhất (đảo ngược)
    if i == -1:
        return arr[::-1]
    
    # Bước 2: Tìm chỉ số j từ phải sang trái sao cho arr[j] > arr[i]
    j = n - 1
    while arr[j] <= arr[i]:
        j -= 1
        
    # Bước 3: Hoán đổi arr[i] và arr[j]
    arr[i], arr[j] = arr[j], arr[i]
    
    # Bước 4: Đảo ngược đoạn từ i+1 đến cuối để được hoán vị kế tiếp nhỏ nhất
    left = i + 1
    right = n - 1
    while left < right:
        arr[left], arr[right] = arr[right], arr[left]
        left += 1
        right -= 1
        
    return arr

# --- Xử lý đầu vào không cần import sys ---
# Đọc n
n = int(input())
# Đọc dãy hoán vị
arr = list(map(int, input().split()))

# Lấy kết quả và in ra
result = next_permutation(arr)
print(*(result))
```

[CanPlaceFlowers] 
Giả sử bạn là nhân viên của một công ty cây xanh, được giao cho chăm sóc một luống 
đất. Luống đất đó được chia thành nhiều ô đất nhỏ có kích cỡ bằng nhau. Trước khi được 
giao cho bạn chăm sóc, trên luống hoa đã được trồng sẵn một số cây xanh tán rộng. Một 
ngày đầu xuân, công ty gửi đến cho bạn một số cây hoa và yêu cầu bạn trồng vào những 
vị trí còn trống trong luống hoa. Đây là những cây hoa hướng sáng và chúng không thể 
sống được dưới bóng râm của cây khác, đặc biệt là cây tán rộng. Vì vậy, bạn không thể 
trồng chúng tại ô bên cạnh những ô đã được trồng cây xanh tán rộng. Và nếu trồng những 
cây hoa bên cạnh nhau, chúng sẽ tranh giành phân bón và nước và sau cùng, cả hai cây 
sẽ cùng chết. 
 
Viết hàm canPlaceFlowers(flowerbed, k) kiểm tra xem có thể trồng hết số hoa mới được 
cấp vào trong luống đất của bạn hay không. 
 
Hàm nhận đầu vào là một danh sách (list) flowerbed, đánh dấu những vị trí trong luống đất 
đã được trồng cây xanh trước đó (giá trị 1 ứng với vị trí cây đã được trồng và 0 ứng với vị 
trí đất trống) và số lượng hoa được phát thêm k (int). Hàm trả về True nếu có thể trồng hết 
toàn bộ k cây hoa vào luống đất và False trong trường hợp ngược lại. 

```python
def canPlaceFlowers(flowerbed, k):
    count = 0
    n = len(flowerbed)
    
    for i in range(n):
        # Kiểm tra nếu vị trí i đang trống
        if flowerbed[i] == 0:
            # Kiểm tra vị trí bên trái (nếu không phải ô đầu)
            left_empty = (i == 0) or (flowerbed[i - 1] == 0)
            # Kiểm tra vị trí bên phải (nếu không phải ô cuối)
            right_empty = (i == n - 1) or (flowerbed[i + 1] == 0)
            
            if left_empty and right_empty:
                # Trồng cây tại i
                flowerbed[i] = 1
                count += 1
                
    # Kiểm tra xem số cây trồng được có đủ k cây không
    return count >= k

# Ví dụ kiểm tra
bed = [1, 0, 0, 0, 1]
k = 1
print(canPlaceFlowers(bed, k)) # Kết quả: True
```

[Phát hiện đường đi lặp lại của Rô-bốt] 
 
Ban đầu, một con rô-bốt ở vị trí (0, 0) trên một lưới 2D.  
Cho một chuỗi `moves` chứa các thao tác di chuyển của rô-bốt. 
 
Các bước di chuyển được biểu diễn bởi các ký tự:  - 'R' (Phải) - 'L' (Trái) - 'U' (Lên) - 'D' (Xuống) 
 
Viết hàm `def is_path_crossing(moves)`: 
Hàm nhận đầu vào là một chuỗi `moves`. 
Hàm trả về `True` nếu trong quá trình di chuyển, rô-bốt đi vào một ô mà nó đã từng đi qua 
trước đó.  
Hàm trả về `False` nếu rô-bốt không bao giờ đi lặp lại (luôn đi vào ô mới). 
 
Lưu ý: Vị trí ban đầu (0, 0) được tính là vị trí đầu tiên đã đi qua.

```python
def is_path_crossing(moves):
    # Khởi tạo vị trí ban đầu
    x, y = 0, 0
    # Sử dụng set để lưu các điểm đã qua, bắt đầu bằng (0, 0)
    visited = {(0, 0)}
    
    for move in moves:
        if move == 'R':
            x += 1
        elif move == 'L':
            x -= 1
        elif move == 'U':
            y += 1
        elif move == 'D':
            y -= 1
            
        # Kiểm tra điểm mới có trong set chưa
        if (x, y) in visited:
            return True
        
        # Thêm điểm mới vào set
        visited.add((x, y))
        
    return False

# Ví dụ kiểm tra
print(is_path_crossing("UURD"))  # Kết quả: False
print(is_path_crossing("UURDL")) # Kết quả: True (Rô-bốt quay lại (0, 0))
```

Viết hàm `def Median(nums):` 
Hàm nhận đầu vào là một danh sách (list) `nums` được đảm bảo có đúng 5 số nguyên. 
 
Yêu cầu: 
Tìm và trả về số trung vị (số lớn thứ ba) của 5 số này. 
 
Ràng buộc: 
1. Chỉ được sử dụng tối đa 6 phép so sánh (ví dụ: <, >, <=, >=). 
2. KHÔNG ĐƯỢC DÙNG bất kỳ hàm sắp xếp nào (như `sorted()`, `list.sort()`). 
 
Chú ý: Bạn chỉ cần viết hàm `Median` theo yêu cầu.  
Phần đọc input và gọi hàm (hàm main) đã có sẵn. 

```python
def Median(nums):
    """
    Tìm trung vị của 5 số với tối đa 6 phép so sánh.
    Không dùng hàm sắp xếp có sẵn.
    """
    a, b, c, d, e = nums[0], nums[1], nums[2], nums[3], nums[4]

    # Bước 1: Sắp xếp cặp đầu tiên (a < b)
    if a > b: a, b = b, a
    
    # Bước 2: Sắp xếp cặp thứ hai (c < d)
    if c > d: c, d = d, c
    
    # Bước 3: Đảm bảo a < c
    if a > c: a, c = c, a; b, d = d, b
    
    # Bước 4, 5, 6: Xác định vị trí của e để tìm trung vị
    if e < c:
        # e nằm bên trái c. So sánh e với b
        if e > b: return e  # Kết quả là e
        else: return b      # Kết quả là b
    else:
        # e nằm bên phải c. So sánh e với d
        if e < d: return e  # Kết quả là e
        else: return d      # Kết quả là d

# --- Phần Test Case ---
def run_tests():
    # Các bộ test kiểm tra độ chính xác
    test_cases = [
        ([3, 1, 5, 2, 4], 3),      # Trung bình
        ([1, 2, 3, 4, 5], 3),      # Đã sắp xếp tăng
        ([5, 4, 3, 2, 1], 3),      # Đã sắp xếp giảm
        ([1, 1, 2, 2, 3], 2),      # Có số trùng lặp
        ([-10, -5, 0, 5, 10], 0),  # Số âm và dương
        ([2, 2, 2, 2, 2], 2)       # Tất cả giống nhau
    ]
    
    print("Đang chạy kiểm thử...")
    for i, (nums, expected) in enumerate(test_cases, 1):
        result = Median(nums)
        status = "PASSED" if result == expected else f"FAILED (Expected {expected}, Got {result})"
        print(f"Test Case {i}: Input {nums} -> {status}")

# Chạy kiểm thử
if __name__ == "__main__":
    run_tests()
```

[Tìm bán kính lò sưởi nhỏ nhất] 
 
Mùa đông đang đến! Bạn được cung cấp vị trí của các ngôi nhà và các lò sưởi trên một 
đường thẳng. 
Nhiệm vụ của bạn là tìm bán kính r nhỏ nhất cho các lò sưởi sao cho mọi ngôi nhà đều 
nằm trong phạm vi sưởi ấm của ít nhất một lò sưởi. 
 
Tất cả các lò sưởi đều có cùng một bán kính r. 
Một ngôi nhà tại vị trí h được sưởi ấm bởi một lò sưởi tại vị trí t nếu |h - t| <= r. 
 
Viết hàm `def findRadius(houses, heaters):` 
Hàm nhận đầu vào là hai danh sách (list) các số nguyên: - `houses`: chứa vị trí của các ngôi nhà. - `heaters`: chứa vị trí của các lò sưởi. 
Hàm trả về một số nguyên là bán kính r nhỏ nhất cần thiết. 
 
Lưu ý: - Số lượng nhà và lò sưởi là một số không âm và không vượt quá 25000. - Vị trí của các ngôi nhà và lò sưởi là một số không âm và không vượt quá 10^9. 

```python
def findRadius(houses, heaters):
    houses.sort()
    heaters.sort()
    
    i = 0
    max_radius = 0
    
    for house in houses:
        # Tìm lò sưởi phù hợp nhất cho ngôi nhà này bằng con trỏ i
        # Lò sưởi hiện tại (heaters[i]) là lò sưởi ở phía trước hoặc tại vị trí nhà
        # Ta tăng i chừng nào lò sưởi tiếp theo vẫn gần ngôi nhà hơn
        while i < len(heaters) - 1 and abs(heaters[i + 1] - house) <= abs(heaters[i] - house):
            i += 1
        
        # Khoảng cách nhỏ nhất của ngôi nhà này đến lò sưởi gần nhất
        current_dist = abs(heaters[i] - house)
        
        # Bán kính lò sưởi phải đủ bao phủ ngôi nhà xa nhất
        max_radius = max(max_radius, current_dist)
        
    return max_radius

# Ví dụ kiểm tra:
print(findRadius([1, 2, 3], [2]))      # Output: 1
print(findRadius([1, 2, 3, 4], [1, 4])) # Output: 1
```

Viết hàm power(base, exp) tính base^exp bằng đệ quy chia để trị.  
 
Yêu cầu:  - Không dùng vòng lặp  - Không dùng hàm có sẵn 

```python
def power(base, exp):
    if exp == 0:
        return 1
    half = power(base, exp // 2)
    if exp % 2 == 0:
        return half * half
    else:
        return base * half * half
print(power(2,10))
```

```python
def find(nums, x):
    countS = {}
    for num in nums:
        counts[num] = counts.get(num, 0) + 1
    result = [num for num, freq in counts.items() if freq > x]
    if not result
        return "INVALID"
    else:
        return result

try:
    nums = list(map(int, input().split()))
    s = int(input())
    output = find(nums, x)
    print(output)
except ValueError:
    print("loi")
```

VIết haàm kiểm tra dãy số nguyên vào từ bàn phím có phải dãy tăng nghiêm ngặt (giảm 
nghiêm ngặt) hay không? 

```python
def check_sequence(nums):
    if len(nums) < 2:
        return "Tăng nghiêm ngặt và Giảm nghiêm ngặt (Dãy quá ngắn)"
    is_increasing = True
    is_decreasing = True
    
    for i in range(len(nums) - 1):
        if nums[i+1] <= nums[i]:
            is_increasing = False
        if nums[i+1] >= nums[i]:
            is_decreasing = False
    if is_increasing:
        return "dãy tăng nghiêm ngặt"
    elif is_decreasing:
        return "dãy giảm nghiêm ngặt"
    else:
        return "dãy không tăng cũng không giảm nghiêm ngặt"
try:
    data = list(map(int, input().split()))
    print(check_sequence(data))
except ValueError:
    print("loi")
```

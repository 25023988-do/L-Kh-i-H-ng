W6A1 	Nhập vào một dòng gồm các số nguyên. Hãy loại bỏ phần tử trùng nhưng giữ nguyên thứ tự xuất hiện đầu tiên, rồi in ra list kết quả. Yêu cầu in ra đúng dạng list Python, ví dụ [1, 2, 3] 

```python
# Nhập dòng số nguyên, chuyển thành list
nums = list(map(int, input().split()))
result = []
seen = set()

for x in nums:
    if x not in seen:
        result.append(x)
        seen.add(x)

print(result) # In đúng dạng list Python
```

W6A4 	Nhập một dòng gồm các cặp key:value cách nhau bởi dấu cách. Tạo dictionary ánh xạ mỗi key -> list các value theo thứ tự xuất hiện. Nếu key lặp lại, thêm vào list của key đó 

```python
# Ví dụ nhập: a:1 b:2 a:3
data = input().split()
d = {}

for item in data:
    key, value = item.split(':')
    if key not in d:
        d[key] = []
    d[key].append(value)

print(d) [cite: 9]
```

W6A9 	Viết chương trình nhập vào hai dictionary. Sau đó gộp lại thành một dictionary mới trong đó nếu một key xuất hiện trong cả hai dict, giá trị sẽ được cộng lại, nếu chỉ có trong một dict thì giữ nguyên. In ra dictionary mới theo thứ tự từ điển tăng dần của key.  

```python
# Nhập dict: k1:v1 k2:v2
def get_dict():
    d = {}
    for item in input().split():
        k, v = item.split(':')
        d[k] = int(v)
    return d

dict1 = get_dict()
dict2 = get_dict()

new_dict = dict1.copy()
for k, v in dict2.items():
    new_dict[k] = new_dict.get(k, 0) + v

# In theo thứ tự từ điển của key
sorted_dict = {k: new_dict[k] for k in sorted(new_dict)}
print(sorted_dict) [cite: 9]
```

W6A10 	Viết chương trình nhập vào một danh sách các số nguyên, và số nguyên dương k. In ra list gồm các cặp số khác nhau có tổng bằng k. Mỗi cặp là một tuple, hai số trong tuple có thể giống nhau, thứ tự xuất hiện trong danh sách của số thứ nhất nhỏ hơn thứ tự của số thứ hai. Các tuple trong list được sắp xếp theo thứ tự tăng dần của phần tử đầu tiên trong tuple. 

```python
nums = list(map(int, input().split()))
k = int(input())
result = []

n = len(nums)
for i in range(n):
    for j in range(i + 1, n):
        if nums[i] + nums[j] == k:
            result.append(tuple(sorted((nums[i], nums[j])))) # Tùy cách hiểu "thứ tự ban đầu"

# Yêu cầu: sắp xếp list kết quả theo phần tử đầu của tuple
result = sorted(list(set(result))) # Loại trùng nếu cần và sắp xếp
print(result) [cite: 9]
```

W6A12 	[Tìm phần tử xuất hiện nhiều nhất] 
Nhập vào một list các số nguyên. Hãy tìm số xuất hiện nhiều lần nhất trong list. Nếu có nhiều số cùng có số lần xuất hiện nhiều nhất, hãy in ra số nhỏ nhất trong số đó. 


```python
nums = list(map(int, input().split()))
counts = {}
for x in nums:
    counts[x] = counts.get(x, 0) + 1

max_freq = max(counts.values())
# Lọc các số có tần suất max, rồi lấy số nhỏ nhất
candidates = [k for k, v in counts.items() if v == max_freq]
print(min(candidates)) [cite: 9]
```

W6A13 	[Đảo ngược key và value của dictionary] 
Nhập vào một dictionary có value là duy nhất (không trùng nhau). Hãy tạo một dictionary mới bằng cách đảo ngược key và value của dictionary ban đầu. In ra dictionary mới. 


```python
# Giả định nhập dạng k:v
data = input().split()
d = {item.split(':')[0]: item.split(':')[1] for item in data}

reversed_dict = {v: k for k, v in d.items()}
print(reversed_dict) [cite: 9]
```

W6A15 	[Lọc dictionary theo giá trị] 
Nhập vào một dictionary có key là chuỗi và value là số nguyên, và một số nguyên k ở dòng tiếp theo. Hãy tạo một dictionary mới chỉ chứa các cặp key-value từ dictionary ban đầu mà có value lớn hơn k. In ra dictionary mới. 


```python
# Nhập dict và k
data = input().split()
d = {item.split(':')[0]: int(item.split(':')[1]) for item in data}
k = int(input())

filtered_dict = {key: val for key, val in d.items() if val > k}
print(filtered_dict) [cite: 10]
```

1 	W6A1 	Nhập vào một dòng gồm các số nguyên. Hãy loại bỏ phần tử trùng nhưng giữ nguyên thứ tự xuất hiện đầu tiên, rồi in ra list kết quả. Yêu cầu in ra đúng dạng list Python, ví dụ [1, 2, 3] 


```python

import sys

def w6a1():
    line = sys.stdin.read().strip()
    if not line: return
    nums = list(map(int, line.split()))
    result = []
    seen = set()
    for x in nums:
        if x not in seen:
            result.append(x)
            seen.add(x)
    print(result)

w6a1()
```

2 	W6A2 	Nhập một dòng số nguyên. In ra list tổng tích lũy (phần tử i là tổng từ đầu tới vị trí i), giữ nguyên thứ tự 	


```python

def w6a2():
    line = input().strip()
    if not line: return
    nums = list(map(int, line.split()))
    res = []
    curr = 0
    for x in nums:
        curr += x
        res.append(curr)
    print(res)
```

3 	W6A3 	Nhập một tuple số nguyên (một dòng) và số nguyên k (dòng kế). In ra tuple sau khi xoay trái k vị trí. (Giữ nguyên thứ tự tương đối sau xoay; không sắp xếp.)


```python

def w6a3():
    try:
        data = input().split()
        if not data: return
        tup = tuple(map(int, data))
        k = int(input())
        n = len(tup)
        k = k % n if n > 0 else 0
        print(tup[k:] + tup[:k])
    except EOFError: pass
```

4 	W6A4 	Nhập một dòng gồm các cặp key:value cách nhau bởi dấu cách. Tạo dictionary ánh xạ mỗi key -> list các value theo thứ tự xuất hiện. Nếu key lặp lại, thêm vào list của key đó 	


```python

def w6a4():
    items = input().split()
    d = {}
    for item in items:
        if ':' in item:
            k, v = item.split(':', 1)
            if k not in d: d[k] = []
            d[k].append(v)
    print(d)
```

5 	W6A5 	Nhập một list số nguyên, In ra dictionary {'positives': ..., 'negatives': ..., 'zeros': ...} (đếm số >0, <0, =0) 

```python

def w6a5():
    nums = list(map(int, input().split()))
    res = {'positives': 0, 'negatives': 0, 'zeros': 0}
    for x in nums:
        if x > 0: res['positives'] += 1
        elif x < 0: res['negatives'] += 1
        else: res['zeros'] += 1
    print(res)
```

9 	W6A9 	Viết chương trình nhập vào hai dictionary. Sau đó gộp lại thành một dictionary mới trong đó nếu một key xuất hiện trong cả hai dict, giá trị sẽ được cộng lại, nếu chỉ có trong một dict thì giữ nguyên. In ra dictionary mới theo thứ tự từ điển tăng dần của key.  	


```python

import ast

def parse_dict(s):
    return ast.literal_eval(s)

def w6a9():
    try:
        d1 = parse_dict(input())
        d2 = parse_dict(input())
        res = d1.copy()
        for k, v in d2.items():
            res[k] = res.get(k, 0) + v
        print({k: res[k] for k in sorted(res)})
    except: pass
```

10 	W6A10 	Viết chương trình nhập vào một danh sách các số nguyên, và số nguyên dương k. In ra list gồm các cặp số khác nhau có tổng bằng k. Mỗi cặp là một tuple, hai số trong tuple có thể giống nhau, thứ tự xuất hiện trong danh sách của số thứ nhất nhỏ hơn thứ tự của số thứ hai. Các tuple trong list được sắp xếp theo thứ tự tăng dần của phần tử đầu tiên trong tuple. 


```python

def w6a10():
    nums = list(map(int, input().split()))
    k = int(input())
    res = []
    for i in range(len(nums)):
        for j in range(i + 1, len(nums)):
            if nums[i] + nums[j] == k:
                res.append((nums[i], nums[j]))
    res.sort()
    print(res)
```

11 	W6A11 	[Tách số chẵn và lẻ từ tuple] 
Nhập vào một tuple chứa các số nguyên. Hãy tạo ra hai tuple mới: một chứa tất cả các số chẵn và một chứa tất cả các số lẻ từ tuple ban đầu, giữ nguyên thứ tự tương đối. In ra hai tuple này trên hai dòng riêng biệt. 	

```python

def w6a11():
    tup = tuple(map(int, input().split()))
    even = tuple(x for x in tup if x % 2 == 0)
    odd = tuple(x for x in tup if x % 2 != 0)
    print(even)
    print(odd)
```

12 	W6A12 	[Tìm phần tử xuất hiện nhiều nhất] 
Nhập vào một list các số nguyên. Hãy tìm số xuất hiện nhiều lần nhất trong list. Nếu có nhiều số cùng có số lần xuất hiện nhiều nhất, hãy in ra số nhỏ nhất trong số đó. 

```python

def w6a12():
    nums = list(map(int, input().split()))
    if not nums: return
    freq = {}
    for x in nums: freq[x] = freq.get(x, 0) + 1
    max_f = max(freq.values())
    candidates = [k for k, v in freq.items() if v == max_f]
    print(min(candidates))
```

13 	W6A13 	[Đảo ngược key và value của dictionary] 
Nhập vào một dictionary có value là duy nhất (không trùng nhau). Hãy tạo một dictionary mới bằng cách đảo ngược key và value của dictionary ban đầu. In ra dictionary mới. 	

```python

def w6a13():
    items = input().split()
    d = {}
    for i in range(0, len(items), 2):
        d[items[i]] = items[i+1]
    res = {v: k for k, v in d.items()}
    print(res)
```

14 	W6A14 	[Giao của hai danh sách] 
Nhập vào hai dòng, mỗi dòng là một danh sách các số nguyên. Hãy tìm các phần tử chung (giao) của hai danh sách này và in ra một list mới chứa các phần tử chung đó, không trùng lặp. 	


```python

def w6a14():
    l1 = set(map(int, input().split()))
    l2 = set(map(int, input().split()))
    res = sorted(list(l1 & l2))
    print(res if res else "[]")
```

15 	W6A15 	[Lọc dictionary theo giá trị] 
Nhập vào một dictionary có key là chuỗi và value là số nguyên, và một số nguyên k ở dòng tiếp theo. Hãy tạo một dictionary mới chỉ chứa các cặp keyvalue từ dictionary ban đầu mà có value lớn hơn k. In ra dictionary mới. 
 


```python

def w6a15():
    parts = input().split()
    d = {}
    for i in range(0, len(parts), 2):
        d[parts[i]] = int(parts[i+1])
    k = int(input())
    res = {key: val for key, val in d.items() if val > k}
    print(res)
```

W6A16 	Nhâp vào mảng n hàng, m cột số nguyên. In lại mảng ra màn hình như ma ̣̣ trận hai chiều, định dạng mỗi phần tử 4 vị trí căn phải 

```python
# Nhập n hàng, m cột
n, m = map(int, input().split())
matrix = []
for _ in range(n):
    row = list(map(int, input().split()))
    matrix.append(row)

# In ma trận định dạng 4 vị trí căn phải 
for row in matrix:
    for val in row:
        print(f"{val:4}", end="")
    print()
```

W6A17 	Nhập vào mảng vuông 2 chiều n hàng, n cột, in ra đường chéo chính và đường chéo phụ của mảng 

```python
n = int(input())
matrix = []
for _ in range(n):
    matrix.append(list(map(int, input().split())))

# Đường chéo chính: các phần tử matrix[i][i] 
main_diag = [matrix[i][i] for i in range(n)]
# Đường chéo phụ: các phần tử matrix[i][n-1-i] 
side_diag = [matrix[i][n-1-i] for i in range(n)]

print(*(main_diag))
print(*(side_diag))
```

W6A18 	Nhâp vào mảng n hàng, m cột số nguyên và số nguyên k (n, m, k nhập trên ̣̣ cùng một dòng, cách nhau dấu cách). Tính và in ra tổng các phần tử trên cột k (hàng k)  

```python
n, m, k = map(int, input().split())
matrix = []
for _ in range(n):
    matrix.append(list(map(int, input().split())))

# Tổng hàng k (0-based)
row_sum = sum(matrix[k]) if k < n else 0
# Tổng cột k (0-based)
col_sum = sum(matrix[i][k] for i in range(n)) if k < m else 0

print(f"Tổng hàng {k}: {row_sum}")
print(f"Tổng cột {k}: {col_sum}")
```

W6A20 	Viết hàm đưa vào một list số nguyên và một số nguyên dương k. Hãy tìm và trả về vị trí của phần tử đầu tiên có giá trị k trong list, nếu không có thì trả về -1 

```python
def find_first_k(nums, k):
    try:
        # Tìm vị trí đầu tiên của k, nếu không có trả về -1 
        return nums.index(k)
    except ValueError:
        return -1
```

W6A21 	Nhập một dòng gồm các chuỗi (tách bởi dấu cách). Tạo dictionary ánh xạ chuỗi -> tuple các vị trí (0-based) mà chuỗi đó xuất hiện 

```python
words = input().split()
pos_dict = {}

for index, word in enumerate(words):
    if word not in pos_dict:
        pos_dict[word] = []
    pos_dict[word].append(index)

# Chuyển list vị trí thành tuple 
result = {k: tuple(v) for k, v in pos_dict.items()}
print(result)
```

W6A23 	VIết hàm xóa đi các số <x có trong mảng 

```python
def remove_less_than_x(nums, x):
    # Tạo list mới chỉ chứa các phần tử >= x 
    nums[:] = [n for n in nums if n >= x]
    return nums
```

W6A24 	Viêết hàm đẩy tất cả các số 0 về đầu (về cuối), các số khác vẫn đúng vị trí trước sau: 0 2 3 2 4 0 0 0 9 => 0 0 0 2 3 2 4 9 

```python
def move_zeros(nums, to_end=True):
    zeros = [n for n in nums if n == 0]
    others = [n for n in nums if n != 0]
    if to_end:
        return others + zeros # Đẩy về cuối 
    else:
        return zeros + others # Đẩy về đầu
```

W6A25 	Cho dãy chỉ gồm N giá trị từ 0 - 9 . Kiểm tra xem giá trị nào không có trong mảng 

```python
def find_missing_digits(nums):
    num_set = set(nums)
    missing = [d for d in range(10) if d not in num_set]
    return missing # In ra các giá trị 0-9 không xuất hiện
```

W6A26 	Tính tổng các phần tử cực đại địa phương trong một danh sách, không tính 2 phần tử đầu và cuối 

```python
def sum_local_maxima(nums):
    total = 0
    for i in range(1, len(nums) - 1):
        if nums[i-1] < nums[i] > nums[i+1]:
            total += nums[i]
    return total
```

W6A27 	Cho một dãy số nguyên, Tìm và in ra vị trí các số < x (x là giá trị cho trước). 
nếu không có in ra -1 


```python
def find_positions_less_than_x(nums, x):
    indices = [i for i, val in enumerate(nums) if val < x]
    if not indices:
        return -1 # Nếu không có in ra -1 
    return indices
```

W6A28 	Cho một dãy các số nguyên, kiểm tra xem nó là dãy tăng thực sự hay giảm thực sự, hay vô danh (không tăng, không giảm) 

```python
def check_sequence(nums):
    if len(nums) < 2: return "vô danh"
    
    is_increasing = all(nums[i] < nums[i+1] for i in range(len(nums)-1))
    is_decreasing = all(nums[i] > nums[i+1] for i in range(len(nums)-1))
    
    if is_increasing: return "tăng thực sự" [cite: 12]
    if is_decreasing: return "giảm thực sự" [cite: 12]
    return "vô danh" [cite: 12]
```

Bài 4: Viết chương trình in ra hình tam giác cân có độ cao h được nhập từ bàn phím 
Ví dụ: Ta nhập h = 4: 
   * 
 *   * 
*     * 
******* 

```python
h = int(input("Nhập độ cao h: "))

for i in range(1, h + 1):
    # Khoảng trắng lề trái
    print(" " * (h - i), end="")
    
    if i == 1:
        print("*")
    elif i == h:
        print("*" * (2 * h - 1))
    else:
        # In dấu sao đầu, khoảng trống giữa, và dấu sao cuối
        print("*" + " " * (2 * i - 3) + "*")
```

Bài 9: Nhập vào số nguyên dương a, kiểm tra xem a có phải là số chính phương hay 
không? Số chính phương là số có căn bậc hai là một số nguyên VD: 1, 4, 9, 16, 25…. 

```python
import math
a = int(input("Nhập a: "))
can = math.isqrt(a) # Lấy phần nguyên căn bậc hai
if can * can == a:
    print(f"{a} là số chính phương")
else:
    print(f"{a} không phải số chính phương")
```

Bài 10: Nhập vào số nguyên dương a, kiểm tra xem a có phải là số hoàn hảo hay không. 
Số hoàn hảo là số có tổng các ước (ước nhỏ hơn nó) bằng chính nó, VD: số 6, 28… 

```python
a = int(input("Nhập a: "))
tong_uoc = 0
for i in range(1, a):
    if a % i == 0:
        tong_uoc += i

if tong_uoc == a:
    print(f"{a} là số hoàn hảo")
else:
    print(f"{a} không phải là số hoàn hảo")
```

Bài 12: Nhập số nguyên dương n, Cho S(k) = 1 + 2 + 3 + … + k. Tìm k sao cho S(k) 
lớn nhất nhưng nhỏ hơn n

```python
n = int(input("Nhập số nguyên dương n: "))
tong = 0
k = 0

# Kiểm tra xem nếu cộng thêm k+1 thì có còn nhỏ hơn n không
while tong + (k + 1) < n:
    k += 1
    tong += k

print(f"Số k lớn nhất là: {k} (với tổng S = {tong} < {n})")
```

Bài 13: Nhập vào số nguyên dương A, tìm n nhỏ nhất sao cho 
1 + 1/2 + 1/3 + 1/4 + ... + 1/n > A

```python
A = float(input("Nhập số nguyên dương A: "))
s = 0
n = 0

while s <= A:
    n += 1
    s += 1/n

print(f"Số n nhỏ nhất là: {n}")
print(f"Tổng thực tế đạt được: {s:.4f}")
```

Bài 14: Nhập vào một dãy số nguyên, dừng nhập khi người dùng nhập -1. Sau khi nhập 
xong, in số lớn nhất, số nhỏ nhất trong những số vừa nhập. KHÔNG ĐƯỢC dùng mảng

```python
max_val = None
min_val = None

print("Nhập dãy số (nhập -1 để dừng):")
while True:
    so = int(input())
    if so == -1:
        break
    
    # Nếu là số đầu tiên, gán nó làm max và min
    if max_val is None:
        max_val = so
        min_val = so
    else:
        # Cập nhật max/min nếu số mới 'xịn' hơn
        if so > max_val:
            max_val = so
        if so < min_val:
            min_val = so

if max_val is not None:
    print(f"Số lớn nhất trong dãy: {max_val}")
    print(f"Số nhỏ nhất trong dãy: {min_val}")
else:
    print("Bạn chưa nhập số nào cả!")
```

Bài 15: Nhập vào số nguyên dương n, đếm xem n có bao nhiêu chữ số. Làm bằng nhiều 
cách nhất có thể, ko dùng mảng. 

```python
n = int(input("Nhập n: "))
dem = 0
temp = n
while temp > 0:
    temp //= 10
    dem += 1
print(f"Cách 1: {dem} chữ số")
```

Bài 16: Nhập vào số nguyên dương n, đếm xem n có bao nhiêu chữ số chẵn, bao nhiêu 
chữ số lẻ

```python
n = int(input("Nhập vào số nguyên dương n: "))

chan = 0
le = 0
temp = n # Dùng biến tạm để không làm mất giá trị gốc của n

# Trường hợp đặc biệt nếu n = 0
if temp == 0:
    chan = 1
else:
    while temp > 0:
        chu_so = temp % 10  # Lấy chữ số cuối cùng
        if chu_so % 2 == 0:
            chan += 1       # Tăng biến đếm chẵn
        else:
            le += 1         # Tăng biến đếm lẻ
        temp //= 10         # Loại bỏ chữ số cuối cùng đã xét
    
print(f"Số chữ số chẵn là: {chan}")
print(f"Số chữ số lẻ là: {le}")
```

Bài 17: Nhập vào số nguyên dương n, tính tổng các chữ số của n

```python
n = int(input("Nhập vào số nguyên dương n: "))

tong = 0
temp = n

while temp > 0:
    chu_so = temp % 10  # Lấy chữ số cuối cùng
    tong += chu_so      # Cộng dồn vào biến tổng
    temp //= 10         # Loại bỏ chữ số cuối cùng để xét số tiếp theo

print(f"Tổng các chữ số của {n} là: {tong}")
```

Bài 18: Nhập vào một số nguyên dương n, kiểm tra xem n có phải là số dạng 3^k hay 
không

```python
n = int(input("Nhập n: "))
is_3k = True

if n <= 0:
    is_3k = False
else:
    temp = n
    while temp > 1:
        if temp % 3 != 0:
            is_3k = False
            break
        temp //= 3

if is_3k:
    print(f"{n} có dạng 3^k")
else:
    print(f"{n} không có dạng 3^k")
```

Bài 19: Dãy số fibonacci là dãy số được định nghĩa như sau: 1, 1, 2, 3, 5, 8, 13,... với số 
kế tiếp sẽ bằng tổng hai số trước đó. Nhập vào số nguyên dương A, hãy tìm số trong 
dãy số fibonacci lớn nhất nhưng không vượt quá A 

```python
A = int(input("Nhập A: "))
f1, f2 = 1, 1
f_max = 1

while f2 <= A:
    f_max = f2
    # Tính số tiếp theo: f_next = f1 + f2
    f1, f2 = f2, f1 + f2

print(f"Số Fibonacci lớn nhất <= {A} là: {f_max}")
```

Bài 20: Nhập vào số nguyên dương n, đếm xem n có bao nhiêu chữ số là số nguyên tố.

```python
n = int(input("Nhập n: "))
dem_nt = 0
temp = n

while temp > 0:
    chu_so = temp % 10
    if chu_so in [2, 3, 5, 7]:
        dem_nt += 1
    temp //= 10

print(f"Số lượng chữ số là số nguyên tố: {dem_nt}")
```

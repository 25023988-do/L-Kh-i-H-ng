Cho một mảng số nguyên không âm. Gọi 
mảng đó là "mảng đẹp" nếu mỗi phần tử 
x trong mảng xuất hiện đúng x lần. 
Kiểm tra xem mảng có phải là mảng đẹp 
hay không. Nếu không hãy xóa đi y số (y 
là nhỏ nhất có thể) để mảng còn lại thành 
mảng đẹp. 

```python
def giai_quyet(arr):
    # 1. Đếm số lần xuất hiện của mỗi số
    dem = {}
    for x in arr:
        dem[x] = dem.get(x, 0) + 1
    
    # 2. Tính số lượng cần xóa
    so_can_xoa = 0
    for x, count in dem.items():
        if x == 0:
            so_can_xoa += count  # Số 0 luôn phải xóa hết
        elif count >= x:
            so_can_xoa += (count - x) # Xóa phần thừa
        else:
            so_can_xoa += count # Không đủ số lượng, xóa sạch
            
    # 3. Trả về kết quả
    if so_can_xoa == 0:
        return "YES"
    else:
        return f"NO {so_can_xoa}"

# Ví dụ chạy thử:
print(giai_quyet([1, 2, 2, 3, 3, 3]))        # YES
print(giai_quyet([1, 2, 3, 4, 5, 6, 7]))     # NO 6
```

W7A17 
Tìm phần tử xuất hiện nhiều hơn n/2 lần 
trong mảng (nếu có) với độ phức tạp thời 
gian là O(n), độ phức tạp bộ nhớ là O(1), - Boyer–Moore. Nếu không có phần tử 
nào thỏa mãn thì không in gì 

```python
def tim_phan_tu_da_so(arr):
    # Bước 1: Tìm ứng cử viên (Boyer-Moore)
    candidate = None
    count = 0
    
    for x in arr:
        if count == 0:
            candidate = x
            count = 1
        elif x == candidate:
            count += 1
        else:
            count -= 1
            
    # Bước 2: Kiểm tra lại xem ứng cử viên có thực sự > n/2 không
    # (Vì thuật toán luôn trả về một số nếu mảng không trống)
    if arr.count(candidate) > len(arr) / 2:
        return candidate
    return None

# Chạy thử với dữ liệu từ hình ảnh
test_cases = [
    [1, 2, 2, 3, 2, 2],       # Kết quả: 2
    [1, 1, 1, 3, 3, 3],       # Không in gì
    [1, 2, 4, 2, 1, 1, 1],    # Kết quả: 1
    [2, 2, 2, 6, 6, 6, 6],    # Kết quả: 6
    [1, 8, 8, 8, 9, 9]        # Không in gì
]

for case in test_cases:
    res = tim_phan_tu_da_so(case)
    if res is not None:
        print(res)
```

W7A18 
Sắp xếp có điều kiện: Số chẵn đứng 
trước, số lẻ đứng sau, chẵn tăng dần, lẻ 
giảm dần 

```python
def sap_xep_co_dieu_kien(arr):
    # 1. Tách chẵn và lẻ
    chan = [x for x in arr if x % 2 == 0]
    le = [x for x in arr if x % 2 != 0]
    
    # 2. Sắp xếp chẵn tăng dần
    chan.sort()
    
    # 3. Sắp xếp lẻ giảm dần
    le.sort(reverse=True)
    
    # 4. Gộp và trả về kết quả
    return chan + le

# Chạy thử với các ví dụ trong image_d18777.png
test_cases = [
    [1, 2, 3, 4, 5, 6],
    [32, 4, 15, 123, 24, 12, 8, 10, 9],
    [3, 4, 2, 6, 7, 8, 9, 10, 5, 7, 2]
]

for i, case in enumerate(test_cases, 1):
    ket_qua = sap_xep_co_dieu_kien(case)
    print(f"Ví dụ {i}: {' '.join(map(str, ket_qua))}")
```

W7A19 
Merge 2 mảng đã sắp xếp (có thể tăng 
dần, có thể giảm dần) mà không dùng 
bất cứ thuật toán sắp xếp nào thành một 
mảng cũng có cùng thứ tự sắp xếp như 2 
mảng ban đầu 

```python
def merge_arrays(arr1, arr2):
    # Kiểm tra xem mảng là tăng dần hay giảm dần
    # (Giả định mảng có ít nhất 2 phần tử để kiểm tra, hoặc mặc định là tăng dần)
    is_descending = len(arr1) > 1 and arr1[0] > arr1[-1]
    
    result = []
    i, j = 0, 0
    
    while i < len(arr1) and j < len(arr2):
        # Điều kiện so sánh dựa trên thứ tự sắp xếp
        if is_descending:
            condition = arr1[i] >= arr2[j]
        else:
            condition = arr1[i] <= arr2[j]
            
        if condition:
            result.append(arr1[i])
            i += 1
        else:
            result.append(arr2[j])
            j += 1
            
    # Thêm các phần tử còn lại của mảng chưa duyệt hết
    result.extend(arr1[i:])
    result.extend(arr2[j:])
    
    return result

# Chạy thử với dữ liệu từ image_d186bb.png
vi_du_1 = merge_arrays([1, 2, 3, 4, 5], [0, 23, 24, 100, 1000])
vi_du_3 = merge_arrays([9, 8, 7, 6, 5, 1, 0], [6, 5, 4, 3, 2, 1])

print("Ví dụ 1 (Tăng dần):", vi_du_1)
print("Ví dụ 3 (Giảm dần):", vi_du_3)
```

W7A20 
Cho dãy N số nguyên dương, tính tổng 
các số có k chữ số. Dòng đầu ghi N số 
nguyên dương, dòng sau ghi số nguyên 
dương k

```python
def xoay_mang_sang_trai(arr, k):
    n = len(arr)
    # Nếu k lớn hơn độ dài mảng, ta lấy k chia dư cho n
    k = k % n 
    
    # Cắt mảng và ghép lại
    ket_qua = arr[k:] + arr[:k]
    return ket_qua

# Chạy thử với dữ liệu từ hình ảnh
test_cases = [
    ([1, 2, 3, 4, 5], 2),    # Xoay 2 lần
    ([1, 2, 3, 4, 5], 10),   # Xoay 10 lần (vẫn là chính nó vì 10 chia hết cho 5)
    ([1, 2, 3, 4, 5], 6),    # Xoay 6 lần (tương đương xoay 1 lần)
    ([1, 2, 3, 4, 5, 6, 7], 3)
]

for arr, k in test_cases:
    print(f"Mảng {arr}, k={k} -> {xoay_mang_sang_trai(arr, k)}")
```

W7A21 
Cho N là ngày, T là tháng năm 2020, 
Tính xem đó là ngày thứ bao nhiêu của 
năm đó nếu đó là ngảy hợp lệ 

```python
def tinh_ngay_thu_bao_nhieu(n, t):
    # Số ngày trong các tháng của năm nhuận 2020
    ngay_trong_thang = [31, 29, 31, 30, 31, 30, 31, 31, 30, 31, 30, 31]
    
    # 1. Kiểm tra tháng hợp lệ
    if t < 1 or t > 12:
        return "invalid"
    
    # 2. Kiểm tra ngày hợp lệ trong tháng đó
    if n < 1 or n > ngay_trong_thang[t-1]:
        return "invalid"
    
    # 3. Tính tổng số ngày
    # sum(ngay_trong_thang[:t-1]) cộng tất cả số ngày của các tháng trước đó
    ngay_thu = sum(ngay_trong_thang[:t-1]) + n
    return ngay_thu

# Chạy thử với các ví dụ trong image_d18337.png
test_cases = [
    (2, 1),   # Ngày 2 tháng 1
    (31, 1),  # Ngày 31 tháng 1
    (4, 5),   # Ngày 4 tháng 5
    (12, 13), # Tháng 13 (Không hợp lệ)
    (23, 8)   # Ngày 23 tháng 8
]

for n, t in test_cases:
    print(f"Input {n} {t} -> Output: {tinh_ngay_thu_bao_nhieu(n, t)}")
```

W7A22 
Cho N tên sinh viên gồm cả họ và tên, 
tìm xem tên nào hay gặp nhất. Nếu có 
nhiều tên cùng số lần gặp thì lấy tên có 
thứ tự từ điển nhỏ nhất 

```python
class SinhVien:
    def __init__(self, ho_ten, msv, diem_toan, diem_van):
        self.ho_ten = ho_ten
        self.msv = msv
        self.diem_toan = diem_toan
        self.diem_van = diem_van
        # Tính điểm trung bình ngay khi khởi tạo
        self.diem_tb = (diem_toan + diem_van) / 2

    def xep_loai(self):
        if self.diem_tb >= 8:
            return "Giỏi"
        elif self.diem_tb >= 6.5:
            return "Khá"
        elif self.diem_tb >= 5:
            return "Trung bình"
        else:
            return "Yếu"

    def hien_thi(self):
        return f"{self.msv} - {self.ho_ten} | ĐTB: {self.diem_tb:.2f} | Loại: {self.xep_loai()}"

# Dữ liệu mẫu từ hình ảnh (giả định)
danh_sach_sv = [
    SinhVien("Nguyễn Văn A", "SV001", 9, 8),
    SinhVien("Lê Thị B", "SV002", 7, 6),
    SinhVien("Trần Văn C", "SV003", 4, 5)
]

# In danh sách kết quả
print("KẾT QUẢ XẾP LOẠI SINH VIÊN:")
print("-" * 40)
for sv in danh_sach_sv:
    print(sv.hien_thi())
```

Đếm số lượng tên trong xâu kí tự. Tên 
được định nghĩa là xâu có độ dài >=2 và 
có chữ cái đầu viết hoa, các chữ cái khác 
viết thường, chỉ có kí tự cái. 

```python
def so_lon_thu_hai_simple(arr):
    # Loại bỏ các số trùng nhau
    unique_nums = list(set(arr))
    
    # Nếu mảng có ít hơn 2 số khác nhau
    if len(unique_nums) < 2:
        return "Không có số lớn thứ hai"
    
    # Sắp xếp tăng dần
    unique_nums.sort()
    
    # Trả về số áp chót
    return unique_nums[-2]

# Ví dụ từ hình: 1 2 3 4 5 -> Kết quả: 4
print(so_lon_thu_hai_simple([1, 2, 3, 4, 5]))
```

W7A24 
Cho xâu kí tự, xóa các từ giống nhau liền 
kề, chỉ giữ lại từ đầu tiên 

```python
def tim_cap_tong_k(arr, k):
    ket_qua = []
    n = len(arr)
    # Duyệt qua từng cặp (i, j)
    for i in range(n):
        for j in range(i + 1, n):
            if arr[i] + arr[j] == k:
                ket_qua.append((arr[i], arr[j]))
    return ket_qua

# Ví dụ từ hình: mảng [1, 2, 3, 4, 5], k = 5
print(tim_cap_tong_k([1, 2, 3, 4, 5], 5)) 
# Kết quả: [(1, 4), (2, 3)]
```

W7A25 
Tìm số có đúng k chữ số lớn nhất có thể 
trong một số có n chữ số (coi đầu vào là 
một số rất lớn - str). Không được hoán 
đổi vị trí các chữ số

```python
import math

def la_so_nguyen_to(n):
    # Số nguyên tố phải lớn hơn 1
    if n < 2:
        return False
    # Kiểm tra từ 2 đến căn bậc hai của n
    for i in range(2, int(math.sqrt(n)) + 1):
        if n % i == 0:
            return False
    return True

def tim_va_sap_xep_snt(arr):
    # 1. Lọc ra các số nguyên tố
    danh_sach_snt = [x for x in arr if la_so_nguyen_to(x)]
    
    # 2. Loại bỏ trùng lặp (nếu cần) và sắp xếp tăng dần
    # Ở đây mình dùng sorted() để sắp xếp
    ket_qua = sorted(danh_sach_snt)
    
    return ket_qua

# Ví dụ dữ liệu từ hình ảnh
mang_input = [10, 7, 3, 2, 8, 11, 5, 4, 13]
# Kết quả mong đợi: 2 3 5 7 11 13

ket_qua = tim_va_sap_xep_snt(mang_input)
print("Các số nguyên tố tăng dần:", " ".join(map(str, ket_qua)))
```

W7A26 
Tách các số có trong một đoạn văn bản, 
tính tổng và tính trung bình các số đó.  

```python
def dem_so_lan_xuat_hien(arr):
    # Khởi tạo từ điển trống
    counts = {}
    
    # Duyệt qua từng phần tử trong mảng
    for x in arr:
        if x in counts:
            counts[x] += 1 # Nếu đã có trong từ điển thì tăng thêm 1
        else:
            counts[x] = 1  # Nếu chưa có thì đặt là 1
            
    # In kết quả theo định dạng yêu cầu
    for gia_tri, so_lan in counts.items():
        print(f"{gia_tri}: {so_lan}")

# Ví dụ dữ liệu từ hình ảnh
mang_input = [1, 2, 2, 3, 3, 3, 4, 4, 4, 4]

print("Kết quả đếm:")
dem_so_lan_xuat_hien(mang_input)
```

W7A27 
Cho một dãy ngoặc ()(){}{}[][]..., kiểm tra 
xem dãy ngoặc đã cho có hợp lệ hay 
không 

```python
def tim_phan_tu_xuat_hien_nhieu_nhat(arr):
    if not arr:
        return None
    
    # 1. Đếm số lần xuất hiện
    counts = {}
    for x in arr:
        counts[x] = counts.get(x, 0) + 1
    
    # 2. Tìm phần tử có số lần xuất hiện lớn nhất
    max_val = arr[0]
    max_count = 0
    
    for x, count in counts.items():
        if count > max_count:
            max_count = count
            max_val = x
            
    return max_val, max_count

# Ví dụ dữ liệu từ hình ảnh
mang_input = [1, 2, 3, 2, 4, 2, 5, 2]

gia_tri, so_lan = tim_phan_tu_xuat_hien_nhieu_nhat(mang_input)
print(f"Phần tử xuất hiện nhiều nhất là {gia_tri} với {so_lan} lần.")
```

W7A29 
Tìm Dãy con tăng dần liên tục dài nhất 
của một danh sách nhập vào từ bàn 
phím. Lấy dãy đầu tiên nếu có nhiều dãy 
có số phần tử bằng nhau 

```python
def tim_cap_chan_le(arr):
    ket_qua = []
    # Duyệt qua từng số a trong mảng
    for a in arr:
        # Duyệt qua từng số b trong mảng
        for b in arr:
            # Kiểm tra điều kiện: a chẵn và b lẻ
            if a % 2 == 0 and b % 2 != 0:
                ket_qua.append((a, b))
    return ket_qua

# Ví dụ dữ liệu từ hình ảnh (giả định mảng [1, 2, 3, 4])
mang_input = [1, 2, 3, 4]
print("Các cặp (chẵn, lẻ):", tim_cap_chan_le(mang_input))
# Kết quả: [(2, 1), (2, 3), (4, 1), (4, 3)]
```

W7A30 
Tìm Dãy con tăng dần KHÔNG liên tục 
dài nhất của một danh sách nhập vào từ 
bàn phím  

```python
import math

# Bước 1: Hàm kiểm tra số nguyên tố tối ưu
def la_so_nguyen_to(n):
    if n < 2: return False
    for i in range(2, int(math.sqrt(n)) + 1):
        if n % i == 0:
            return False
    return True

# Bước 2 & 3: Tìm các cặp có tổng là số nguyên tố
def tim_cap_tong_nguyen_to(arr):
    n = len(arr)
    ket_qua = []
    
    # Duyệt cặp (i, j) để tránh trùng lặp vị trí
    for i in range(n):
        for j in range(i + 1, n):
            tong = arr[i] + arr[j]
            if la_so_nguyen_to(tong):
                ket_qua.append((arr[i], arr[j]))
    return ket_qua

# Ví dụ chạy thử với dữ liệu
mang_input = [1, 2, 3, 4, 5]
ket_qua = tim_cap_tong_nguyen_to(mang_input)

print("Các cặp có tổng là số nguyên tố:")
for cap in ket_qua:
    print(f"{cap[0]} + {cap[1]} = {sum(cap)} (SNT)")
```

Đổi chỗ đúng 2 chữ số để được số nhỏ 
nhất có thể 

```python
import math

# Bước 1: Hàm tìm Ước chung lớn nhất (GCD)
def tim_ucln(a, b):
    while b:
        a, b = b, a % b
    return a

# Bước 2 & 3: Tìm các cặp nguyên tố cùng nhau
def tim_cap_nguyen_to_cung_nhau(arr):
    n = len(arr)
    ket_qua = []
    
    for i in range(n):
        for j in range(i + 1, n):
            # Kiểm tra nếu GCD của cặp số bằng 1
            if math.gcd(arr[i], arr[j]) == 1:
                ket_qua.append((arr[i], arr[j]))
    return ket_qua

# Ví dụ chạy thử
mang_input = [2, 3, 4, 9, 10]
ket_qua = tim_cap_nguyen_to_cung_nhau(mang_input)

print("Các cặp số nguyên tố cùng nhau:")
for cap in ket_qua:
    print(f"GCD({cap[0]}, {cap[1]}) = 1")
```

W8A16 Yêu cầu 
Xây dựng lớp Shape với phương thức area(). 
Xây dựng lớp Rectangle(w, h) và Circle(r) kế thừa từ Shape, ghi đè area(). 
Lấy π = 3.14. 
 
Input 
Dòng 1: loại hình (RECTANGLE hoặc CIRCLE) 
Dòng 2: tham số tương ứng. 
 
Output 
In ra diện tích với 2 chữ số thập phân. 

```python
import math

# 1. Xây dựng lớp cha Shape
class Shape:
    def area(self):
        pass

# 2. Xây dựng lớp Rectangle kế thừa từ Shape
class Rectangle(Shape):
    def __init__(self, width, height):
        self.width = width
        self.height = height
    
    # Ghi đè phương thức area()
    def area(self):
        return self.width * self.height

# 3. Xây dựng lớp Circle kế thừa từ Shape
class Circle(Shape):
    def __init__(self, radius):
        self.radius = radius
    
    # Ghi đè phương thức area() với pi = 3.14
    def area(self):
        return 3.14 * (self.radius ** 2)

# --- Phần xử lý Input và Output ---

def main():
    try:
        # Đọc loại hình (RECTANGLE hoặc CIRCLE)
        shape_type = input().strip().upper()
        
        # Đọc tham số tương ứng
        params = input().split()
        
        if shape_type == "RECTANGLE":
            w = float(params[0])
            h = float(params[1])
            shape = Rectangle(w, h)
        elif shape_type == "CIRCLE":
            r = float(params[0])
            shape = Circle(r)
        else:
            return

        # In kết quả với 2 chữ số thập phân
        print(f"{shape.area():.2f}")
        
    except EOFError:
        pass

if __name__ == "__main__":
    main()
```

Xây dựng lớp Account(balance) với hai phương thức deposit(x) và get_balance(). 
Xây dựng lớp SavingAccount(balance, rate) kế thừa từ Account, thêm phương thức 
add_interest() để cộng balance*rate vào số dư. 
 
Input 
balance rate 
 
Output 
Số dư sau khi cộng lãi, in 2 chữ số thập phân.

```python
# 1. Xây dựng lớp cha Account
class Account:
    def __init__(self, balance):
        self.balance = float(balance)
    
    def deposit(self, x):
        self.balance += float(x)
    
    def get_balance(self):
        return self.balance

# 2. Xây dựng lớp SavingAccount kế thừa từ Account
class SavingAccount(Account):
    def __init__(self, balance, rate):
        # Sử dụng super() để khởi tạo balance từ lớp cha
        super().__init__(balance)
        self.rate = float(rate)
    
    # Phương thức cộng lãi suất vào số dư
    def add_interest(self):
        interest = self.balance * self.rate
        self.balance += interest

# --- Phần xử lý Input và Output ---

def main():
    try:
        # Đọc balance và rate từ một dòng input
        data = input().split()
        if len(data) == 2:
            initial_balance = float(data[0])
            interest_rate = float(data[1])
            
            # Tạo đối tượng tài khoản tiết kiệm
            my_saving = SavingAccount(initial_balance, interest_rate)
            
            # Gọi phương thức cộng lãi
            my_saving.add_interest()
            
            # In ra số dư cuối cùng với 2 chữ số thập phân
            print(f"{my_saving.get_balance():.2f}")
            
    except (EOFError, ValueError, IndexError):
        pass

if __name__ == "__main__":
    main()
```

W8A22 Yêu cầu 
Xây dựng lớp Matrix2x2(a, b, c, d) biểu diễn ma trận: 
[a b] 
[c d] 
Viết phương thức trace() trả về a + d và determinant() trả về ad - bc. 
 
Input 
Một dòng chứa 4 số nguyên a b c d. 
 
Output 
In ra trace và determinant trên cùng một dòng. 

```python
# Xây dựng lớp Matrix2x2
class Matrix2x2:
    def __init__(self, a, b, c, d):
        # Ma trận có dạng:
        # [a b]
        # [c d]
        self.a = int(a)
        self.b = int(b)
        self.c = int(c)
        self.d = int(d)
    
    # Phương thức tính vết của ma trận (tổng các phần tử đường chéo chính)
    def trace(self):
        return self.a + self.d
    
    # Phương thức tính định thức (determinant)
    def determinant(self):
        return (self.a * self.d) - (self.b * self.c)

# --- Phần xử lý Input và Output ---

def main():
    try:
        # Đọc 4 số nguyên a, b, c, d trên cùng một dòng
        line = input().split()
        if len(line) == 4:
            a, b, c, d = line
            
            # Khởi tạo đối tượng ma trận
            matrix = Matrix2x2(a, b, c, d)
            
            # Lấy kết quả từ các phương thức
            t = matrix.trace()
            det = matrix.determinant()
            
            # In kết quả trên cùng một dòng, cách nhau bởi khoảng trắng
            print(f"{t} {det}")
            
    except (EOFError, ValueError):
        pass

if __name__ == "__main__":
    main()
```

W8A28 Yêu cầu 
Xây dựng lớp Counter có thuộc tính count ban đầu bằng 0 và các phương thức inc(), dec(), 
value(). 
Sau đó thực hiện dãy thao tác trên đối tượng. 
 
Input 
Dòng 1: n 
Dòng 2: n thao tác, mỗi thao tác là INC hoặc DEC, cách nhau bởi dấu cách. 
 
Output 
Giá trị cuối cùng của count.

```python
# 1. Xây dựng lớp Counter
class Counter:
    def __init__(self):
        # Thuộc tính count ban đầu bằng 0
        self.count = 0
    
    # Phương thức tăng giá trị
    def inc(self):
        self.count += 1
    
    # Phương thức giảm giá trị
    def dec(self):
        self.count -= 1
    
    # Phương thức lấy giá trị hiện tại
    def value(self):
        return self.count

# --- Phần xử lý Input và Output ---

def main():
    try:
        # Dòng 1: Đọc số lượng thao tác n
        n = int(input().strip())
        
        # Dòng 2: Đọc danh sách các thao tác (INC hoặc DEC)
        operations = input().split()
        
        # Khởi tạo đối tượng bộ đếm
        my_counter = Counter()
        
        # Thực hiện dãy thao tác dựa trên input
        # Lưu ý: Chỉ thực hiện tối đa n thao tác theo đề bài
        for i in range(min(n, len(operations))):
            op = operations[i].upper()
            if op == "INC":
                my_counter.inc()
            elif op == "DEC":
                my_counter.dec()
        
        # In ra giá trị cuối cùng của count
        print(my_counter.value())
            
    except (EOFError, ValueError):
        pass

if __name__ == "__main__":
    main()
```

Dữ liệu đảm bảo mọi lệnh pop và top đều hợp lệ. 
 
Input 
Dòng 1: n 
n dòng tiếp theo là một trong ba lệnh: 
PUSH x 
POP 
TOP 
 
Output 
Mỗi lệnh TOP in ra giá trị ở đỉnh ngăn xếp trên một dòng.

```python
# 1. Xây dựng lớp Stack
class Stack:
    def __init__(self):
        # Sử dụng một danh sách rỗng để lưu trữ các phần tử
        self.items = []
    
    # Thêm phần tử vào đỉnh ngăn xếp
    def push(self, x):
        self.items.append(x)
    
    # Loại bỏ phần tử ở đỉnh ngăn xếp
    def pop(self):
        if not self.is_empty():
            return self.items.pop()
    
    # Trả về giá trị của phần tử ở đỉnh mà không loại bỏ nó
    def top(self):
        if not self.is_empty():
            return self.items[-1]
    
    # Kiểm tra ngăn xếp có rỗng hay không
    def is_empty(self):
        return len(self.items) == 0

# --- Phần xử lý Input và Output ---

def main():
    try:
        # Đọc số lượng lệnh n
        line1 = input().strip()
        if not line1:
            return
        n = int(line1)
        
        my_stack = Stack()
        
        # Xử lý n dòng tiếp theo
        for _ in range(n):
            command = input().split()
            if not command:
                continue
                
            action = command[0].upper()
            
            if action == "PUSH":
                x = command[1]
                my_stack.push(x)
            elif action == "POP":
                my_stack.pop()
            elif action == "TOP":
                print(my_stack.top())
                
    except EOFError:
        pass

if __name__ == "__main__":
    main()
```

Xây dựng lớp Cylinder (hình trụ) gồm: bán kính và chiều cao (kiểu số thực) với hai phương thức:tính diện tích bề mặt (diện tích toàn phần) và thể tích của hình trụ 
Đầu vào  
Đâu vào từ bàn phím gồm một dòng duy nhất chứa bán kính và chiều cao của hình trụ, cách nhau bởi một dấu cách. 
 
Đầu ra 
In ra màn hình diện tích bề mặt và thể tích của hình trụ (sử dụng lại những hàm vừa viết, π=3.14).  
 
(Kết quả làm tròn đến 2 chữ số sau dấu phẩy). 


```python
class Cylinder:
    def __init__(self, r, h):
        self.bankinh = r
        self.chieucao = h
    def getArea(self):
        return 2*3.14*self.bankinh**2 + 2*3.14*self.bankinh*self.chieucao
    def getVolume(self):
        return self.chieucao*self.bankinh**2*3.14
    def __str__(self):
        return f"{self.bankinh}, {self.chieucao}"
b, c = map(float, input().split())
cy1 = Cylinder(b, c)
print(cy1)
print(f"dien tich toan phan: {cy1.getArea()}")
print(f"the tich: {cy1.getVolume()}")
```

W8A2 	[Date] 
Xây dựng lớp Date biểu diễn ngày tháng năm với hai phương thức kiểm tra xem một ngày tháng nào đó có hợp lệ hay không và phương thức tính ngày kế tiếp của ngày tháng năm nào đó nếu nó là hợp lệ 
 
Nhập vào một ngày tháng năm, (ngày tháng năm biểu diễn dưới dạng dd/mm/yyyy). Nếu ngày tháng năm không tồn tại, in ra dòng chữ INVALID. Nếu là hợp lệ in ra ngày kế tiếp của nó. 
 
Đầu vào  
Đầu vào từ bàn phim chứa dữ liệu ngày, tháng, năm dưới dạng dd/mm/yyyy 
. 
Đầu ra 
In ra màn hình ngày kế tiếp của ngày vừa nhập dưới dạng dd/mm/yyyy 
. Nếu ngày tháng không tồn tại, in ra INVALID 


```python
class Mydate:
    def __init__(self, day, month, year):
        self.day = day
        self.month = month
        self.year = year
    def checkValid(self):
        maxday = -1
        match self.month:
            case 1|3|5|7|10|12:
                maxday = 31
            case 4|6|9|11:
                maxday = 30
            case 2:
                if(self.year %4 == 0 and self.year % 100 != 0) and (self.year % 400 == 0):
                    maxday = 29
                else:
                    maxday = 28
        return 1 <= self.day <= maxday and self.year > 0
    def getNextday(self):
        if(self.checkValid()):
            day1 = Mydate(self.day +1, self.month, self.year)
            if(day1.checkValid()):
                return day1
            day1 = Mydate(1, self.month + 1, self.year)
            if(day1.checkValid()):
                return day1
        return Mydate(1, 1, self.year + 1)
    def __str__(self):
        return f"{self.day:02d}/{self.month:02d}/{self.year}"

d, m, y = map(int, input().split('/'))
day1 = Mydate(d, m, y)
print(day1.checkValid())
print(day1.getNextday())
```

Xây dựng lớp điểm với hai thuộc tính x, y cùng các phương thức kiểm tra xem có phải gốc tọa dộ không, có nằm trên trục tung không, có nằm trên trục hoành không, tính khoảng cách từ điểm đó tới một điểm khác 
 
Đầu vào:  
Nhập vào tọa độ một điểm M bất kỳ 
 
Đầu ra: 
Thông báo M có phải gốc tọa độ không hoặc có nằm trên trục tung hoặc trục hoành không, khoảng cách từ M tới gốc tọa độ là bao nhiêu (lấy 2 chữ số phần thập phân) 


```python

class Point:
    def __init__(self, x, y):
        self.x = x
        self.y = y
    def check(self):
        if self.x == 0 and self.y == 0:
            return f"là gốc tọa độ"

        else:
            return f"không phải là gốc tọa độ"
    def tung(self):
        if self.x == 0 and self.y != 0:
            return f"nằm trên trục tung"
        return ' '
    def hoanh(self):
        if self.y == 0 and self.x != 0:
            return f"nằm trên trục hoành"
        return ''
    def khoang_cach(self, other):
        dx = self.x - other.x
        dy = self.y - other.y
        d = (dx**2 + dy**2)**0.5
        return f'khoang cach: {d:.2f}'
x, y = map(float, input().split())
O = Point(0, 0)
c1 = Point(x, y)
print(c1.check(), c1.tung(), c1.hoanh(), c1.khoang_cach(O))


```

W8A5 	[ShoppingCart] 
Xây dựng lớp ShoppingCart (giỏ hàng) để quản lý các sản phẩm của khách. Dùng danh sách để lưu các sản phẩm, mỗi 	 	 

	sản phẩm là một từ điển có name (tên) và price(giá).  
Với các phương thức sau: 
1.	Khởi tạo với danh sách rỗng 
2.	Thêm vào giỏ hàng một sản phẩm mới 
3.	Xóa khỏi giỏ một sản phẩm nào đó (tìm theo tên để xóa) 
4.	Kiểm tra xem giỏ hàng có rỗng không 
5.	Tính tổng tiền giỏ hàng 
6.	Hiển thị toàn bộ sản phẩm trong giỏ 
7.	Xóa toàn bộ giỏ hàng 
 
Tạo menu cho người dùng nhập lựa chọn truy vấn từ 2 tới 7. Chương trình dừng khi người dùng nhập "yes" cho câu hỏi "Có muốn thoát khỏi chương trình không?" 		


```python
class ShoppingCart:
    # 1. Khởi tạo với danh sách rỗng
    def __init__(self):
        self.cart = []

    # 2. Thêm vào giỏ hàng một sản phẩm mới
    def add_product(self, name, price):
        product = {"name": name, "price": price}
        self.cart.append(product)
        print(f"-> Đã thêm '{name}' vào giỏ hàng.")

    # 3. Xóa khỏi giỏ một sản phẩm nào đó (tìm theo tên để xóa)
    def remove_product(self, name):
        for product in self.cart:
            # So sánh không phân biệt chữ hoa chữ thường bằng .lower()
            if product["name"].lower() == name.lower():
                self.cart.remove(product)
                print(f"-> Đã xóa sản phẩm '{product['name']}' khỏi giỏ hàng.")
                return
        print(f"-> Không tìm thấy sản phẩm nào có tên '{name}' trong giỏ!")

    # 4. Kiểm tra xem giỏ hàng có rỗng không
    def is_empty(self):
        return len(self.cart) == 0

    # 5. Tính tổng tiền giỏ hàng
    def get_total_price(self):
        total = 0
        for product in self.cart:
            total += product["price"]
        return total

    # 6. Hiển thị toàn bộ sản phẩm trong giỏ
    def display_cart(self):
        if self.is_empty():
            print("-> Giỏ hàng hiện đang rỗng.")
            return
        
        print("\n--- DANH SÁCH SẢN PHẨM TRỒNG GIỎ ---")
        for i, product in enumerate(self.cart, start=1):
            print(f"{i}. Tên: {product['name']} | Giá: {product['price']:,}đ")
        print("------------------------------------")

    # 7. Xóa toàn bộ giỏ hàng
    def clear_cart(self):
        self.cart.clear()
        print("-> Đã làm trống toàn bộ giỏ hàng.")


# --- CHƯƠNG TRÌNH CHÍNH (MENU ĐIỀU KHIỂN) ---
if __name__ == "__main__":
    my_cart = ShoppingCart()
    print("=== HỆ THỐNG QUẢN LÝ GIỎ HÀNG ===")

    while True:
        # Hiển thị Menu chức năng từ 2 tới 7
        print("\n" + "="*30)
        print("2. Thêm sản phẩm vào giỏ")
        print("3. Xóa sản phẩm khỏi giỏ (theo tên)")
        print("4. Kiểm tra giỏ hàng có rỗng không")
        print("5. Tính tổng tiền giỏ hàng")
        print("6. Hiển thị tất cả sản phẩm trong giỏ")
        print("7. Xóa toàn bộ giỏ hàng")
        print("8. Thoát chương trình")
        print("="*30)
        
        choice = input("Mời bạn nhập lựa chọn (2-8): ").strip()

        if choice == '2':
            name = input("Nhập tên sản phẩm: ").strip()
            # Đảm bảo người dùng nhập giá trị số hợp lệ
            try:
                price = float(input("Nhập giá sản phẩm: "))
                my_cart.add_product(name, price)
            except ValueError:
                print("Lỗi: Giá sản phẩm phải là một số!")
                
        elif choice == '3':
            name = input("Nhập tên sản phẩm cần xóa: ").strip()
            my_cart.remove_product(name)
            
        elif choice == '4':
            if my_cart.is_empty():
                print("-> Giỏ hàng đang RỖNG.")
            else:
                print(f"-> Giỏ hàng KHÔNG RỖNG (đang có {len(my_cart.cart)} sản phẩm).")
                
        elif choice == '5':
            total = my_cart.get_total_price()
            print(f"-> Tổng tiền các sản phẩm trong giỏ: {total:,}đ")
            
        elif choice == '6':
            my_cart.display_cart()
            
        elif choice == '7':
            confirm = input("Bạn có chắc chắn muốn xóa sạch giỏ hàng không? (y/n): ").strip().lower()
            if confirm == 'y':
                my_cart.clear_cart()
                
        elif choice == '8':
            # Nhánh xử lý thoát chương trình
            exit_choice = input("Có muốn thoát khỏi chương trình không? (Gõ 'yes' để thoát): ").strip()
            if exit_choice == "yes":
                print("Cảm ơn bạn đã mua sắm. Tạm biệt!")
                break
            else:
                print("Tiếp tục mua sắm...")
                
        else:
            print("Lựa chọn không hợp lệ, vui lòng chọn từ 2 đến 8!")
```

W8A6 	Yêu cầu 
Xây dựng lớp Rectangle(w, h) với các phương thức: 
 
area() trả về diện tích. 
 
perimeter() trả về chu vi. 
 
scale(k) phóng to hình chữ nhật theo hệ số k (nhân cả w và h), không trả về gì. 
 
Input 
w h k (số thực dương) 
 
Output area perimeter sau khi phóng to (làm tròn 2 chữ số thập phân). 


```python
class Rectangle:
    def __init__(self, w, h):
        self.w = w
        self.h = h
    def area(self):
        return self.w * self.h 
    def perimeter(self):
        return f"{(2 * (self.w + self.h)):.2f}"
    def scale(self, k):
        self.w *= k
        self.h *= k
    def __str__(self):
        return f"Recctangle(w={self.w}, h={self.h})"
w, h = map(float, input("nhap chieu rong va chieu dai:").split())
k = float(input("nhap he so scale:").strip())
rect = Rectangle(w, h)
rect.scale(k)
print(rect.area())
print(rect.perimeter())
print(rect)
```

W8A7 	Yêu cầu 
Lớp Fraction(num, den) (tử, mẫu nguyên, mẫu ≠ 0) với: 
 
Rút gọn phân số trong __init__ (quy ước mẫu > 0). 
 
Các phương thức add/sub/mul/div trả về phân số đã rút gọn. 
 
__str__() in dạng a/b. 
 
Input 
a b op c d (tương ứng a/b (op) c/d, op thuộc {+,-,*,/}) 
 
Output 
Phân số kết quả đã rút gọn 


```python
class Fraction:
    def __init__(self, num, den):
        if den == 0:
            raise ValueError("mau so khac 0")
        # tim uoc chung lon nhat
        def ucln(num, den):
            while den != 0:
                phan_du = num % den
                num = den
                den = phan_du
            return abs(num)
        self.num = num // ucln(num,den)
        self.den = den // ucln(num,den)
        if self.den < 0:
            self.num = -self.num
            self.den = -self.den
    def add(self, othet):
        new_num = self.num * othet.den + othet.num * self.den
        new_den = self.den * othet.den
        return Fraction(new_num, new_den)
    def sub(self, othet):
        new_num = self.num * othet.den - othet.num * self.den
        new_den = self.den * othet.den
        return Fraction(new_num, new_den)
    def mul(self, othet):
        new_num = self.num * othet.num
        new_den = self.den * othet.den
        return Fraction(new_num, new_den)
    def div(self, othet):
        if othet.num == 0:
            raise ValueError("khong the chia cho 0")
        new_num = self.num * othet.den
        new_den = self.den * othet.num
        return Fraction(new_num, new_den)
    def __str__(self):
        if self.den == 1:
            return f"{self.num}"
        return f"{self.num}/{self.den}"
num1, den1 = map(int, input("nhap tu so va mau so phan so 1:").split())
num2, den2 = map(int, input("nhap tu so va mau so phan so 2:").split())
frac = Fraction(num1, den1)
frac2 = Fraction(num2, den2)
print(f"tong: {frac.add(frac2)}")
print(f"hieu: {frac.sub(frac2)}")
print(f"tich: {frac.mul(frac2)}")
print(f"thuong: {frac.div(frac2)}")



        
```

Lớp BankAccount(owner, balance) với:

Xây dựng lớp Product(name, price). 
Xây dựng lớp DiscountProduct(name, price, percent) kế thừa từ Product, ghi đè phương thức final_price() để trả về giá sau giảm. Công thức: price * (100 - percent) / 100 
 
Input 
name price percent 
Output 
In ra giá cuối cùng với 2 chữ số thập phân. 



```python
class Product:
    def __init__(self, name, price):
        self.name = name
        self.price = price
    def final_price(self):
        return self.price
class DiscountProduct(Product):
    def __init__(self, name, price, percent):
        super().__init__(name, price)
        self.percent = percent
    def final_price(self):
        return self.price * (100 - self.percent) / 100
name = input("nhap ten san pham:").strip()
price = float(input("nhap gia san pham:").strip())
percent = float(input("nhap phan tram giam gia:").strip())
discount_product = DiscountProduct(name, price, percent)
print(f"gia sau khi giam: {discount_product.final_price():.2f}")
```

Xây dựng lớp Counter có thuộc tính count ban đầu bằng 0 và các phương thức inc(), dec(), value(). 
Sau đó thực hiện dãy thao tác trên đối tượng. 
 
Input 
Dòng 1: n 
Dòng 2: n thao tác, mỗi thao tác là INC hoặc DEC, cách nhau bởi dấu cách. 
 
Output 
Giá trị cuối cùng của count. 


```python

class Counter:
    def __init__(self):
        self.count = 0
    def inc(self):
        self.count += 1
    def dec(self):
        self.count -= 1
    def value(self):
        return self.count
n = int(input("nhap so thao tac:").strip())
thap_tac = input("nhap thao tac:").split()
counter = Counter()
for x in thap_tac:
    if x == "inc":
        counter.inc()
    elif x == "dec":
        counter.dec()
print(counter.value())
    
```

Xây dựng lớp Time(h, m, s) với phương thức to_seconds() đổi thời gian sang tổng số giây. 
 
Input h m s 
 
Output 
Tổng số giây. 


```python
class Time:
    def __init__(self, h, m, s):
        self.h = h
        self.m = m
        self.s = s
    def to_seconds(self):
        return self.h * 3600 + self.m *60 + self.s
h, m, s = map(int, input("nhap gio, phut va giay:").split())
time = Time(h, m, s)
print(time.to_seconds())
```

W8A30 	Yêu cầu 
Xây dựng lớp Stack sử dụng danh sách để cài đặt ngăn xếp với các phương thức push(x), pop(), top(). 
Dữ liệu đảm bảo mọi lệnh pop và top đều hợp lệ. 
 
Input Dòng 1: n n dòng tiếp theo là một trong ba lệnh: 
PUSH x 
POP 
TOP 
 
Output 
Mỗi lệnh TOP in ra giá trị ở đỉnh ngăn xếp trên một dòng. 


```python
class Stack:
    def __init__(self):
        self.stack = []
    def push(self, x):
        self.stack.append(x)
    def is_empty(self):
        return len(self.stack) == 0
    def pop(self):
        if not self.is_empty():
            return self.stack.pop()
        else:
            return "Stack is empty"
    def top(self):
        if not self.is_empty():
            return self.stack[-1]
        else:
            return "Stack is empty"
n = int(input().strip())
stack = Stack()

for _ in range(n):
    line = input().split()
    cmd = line[0].upper() # Chuyển thành chữ in hoa để khớp với đề
    
    if cmd == "PUSH":
        x = int(line[1])
        stack.push(x)
    elif cmd == "POP":
        stack.pop()
    elif cmd == "TOP":
        print(stack.top())

```

[PrimeNumber] 
VIết chương trình nhập vào một số tới khi được một số nguyên dương thì thôi. 
Kiểm tra số này có phải là số nguyên tố hay không?

```python
# Vòng lặp đảm bảo nhập đúng số nguyên dương
while True:
    n = int(input("Nhập một số nguyên dương: "))
    if n > 0:
        break
    print("Vui lòng nhập lại (số phải > 0)!")

# Kiểm tra số nguyên tố
is_prime = True
if n < 2:
    is_prime = False
else:
    for i in range(2, int(n**0.5) + 1):
        if n % i == 0:
            is_prime = False
            break

if is_prime:
    print(f"{n} là số nguyên tố.")
else:
    print(f"{n} không phải là số nguyên tố.")
```

.[CountingDigits] 
Viết chương trình nhập vào một số nguyên n và in ra màn hình số chữ số (trừ dấu) 
của số đó. Không dùng kiểu dữ liệu string. 

```python
n = int(input("Nhập một số nguyên n: "))
n_phu = abs(n) # Lấy giá trị tuyệt đối để xử lý cả số âm
dem = 0

if n_phu == 0:
    dem = 1
else:
    while n_phu > 0:
        n_phu //= 10  # Chia lấy phần nguyên cho 10 để bớt đi 1 chữ số
        dem += 1

print(f"Số chữ số là: {dem}")
```

W4A6 
"[SumPrimesInRange] 
Nhập hai số nguyên dương a, b (a ≤ b). Hãy tính tổng các số nguyên tố trong đoạn 
[a, b]. Mỗi dòng là một bộ test (a b)."

```python
def is_prime(n):
    if n < 2: return False
    for i in range(2, int(n**0.5) + 1):
        if n % i == 0: return False
    return True

# Giả sử nhập nhiều bộ test cho đến khi không nhập nữa
try:
    while True:
        line = input().split()
        if not line: break
        a, b = map(int, line)
        tong = sum(i for i in range(a, b + 1) if is_prime(i))
        print(tong)
except EOFError:
    pass
```

W4A7 
"[LargestPrimeFactor] 
Nhập vào số nguyên dương n (n ≥ 2). Hãy in ra ước số nguyên tố lớn nhất của n."

```python
n = int(input("Nhập n: "))
d = 2
max_prime = 1
temp_n = n

while d * d <= temp_n:
    while temp_n % d == 0:
        max_prime = d
        temp_n //= d
    d += 1
if temp_n > 1:
    max_prime = temp_n

print(max_prime)
```

W4A8 
ReverseAddPalindrome] 
Nhập số nguyên dương n. Lặp lại phép toán n = n + reverse(n) cho đến khi n là số 
đối xứng (palindrome). In ra số bước và giá trị palindrome thu được.

```python
def is_palindrome(n):
    return str(n) == str(n)[::-1]

n = int(input("Nhập n: "))
steps = 0
while not is_palindrome(n):
    n = n + int(str(n)[::-1])
    steps += 1

print(f"{steps} {n}")
```

[PerfectSquaresDistinctDigits] 
Nhập vào số nguyên dương n. In ra tất cả các số chính phương ≤ n có các chữ số 
đôi một khác nhau (in theo thứ tự tăng dần, cùng hàng, cách nhau bởi một dấu 
cách). Nếu không có, in ra (no number).

```python
import math

n = int(input("Nhập n: "))
results = []

for i in range(1, int(math.sqrt(n)) + 1):
    square = i * i
    s_square = str(square)
    # Kiểm tra các chữ số có đôi một khác nhau không
    if len(set(s_square)) == len(s_square):
        results.append(s_square)

if results:
    print(" ".join(results))
else:
    print("no number")
```

W4A10 
"[LongestCollatzUpToN] 
Nhập vào số nguyên dương n. Tìm số x trong [1, n] có độ dài dãy Collatz lớn nhất 
và in ra x cùng độ dài dãy tương ứng (x L). Nếu có nhiều số cùng độ dài, chọn số 
nhỏ nhất."

```python
def collatz_len(num):
    count = 1
    while num > 1:
        if num % 2 == 0: num //= 2
        else: num = 3 * num + 1
        count += 1
    return count

n = int(input("Nhập n: "))
max_len = 0
best_x = 0

for x in range(1, n + 1):
    length = collatz_len(x)
    if length > max_len:
        max_len = length
        best_x = x

print(f"{best_x} {max_len}")
```

[ClosePairNumber] 
Hai số nguyên dương tạo thành một cặp số thân thiết khi chúng tuân theo quy luật: 
Số này bằng tổng tất cả các ước của số kia (trừ chính nó) và ngược lại. 
Ví dụ 220 và 284 là một cặp số thân thiết 
220 ngoài chính nó ra, nó còn có 11 ước số khác là: 1, 2, 4, 5, 10, 11, 20, 44, 55, 
110. Tổng của 11 ước này cho chúng ta 284 
284 ngoài bản thân nó, còn có 5 ước số khác là: 1, 2, 4, 71, 142. Tổng của chúng 
vừa đúng bằng 220 
Nhập hai số từ bàn phím và kiểm tra xem chúng có phải là cặp số thân thiết không. 
Nếu có, in ra màn hình "true", ngược lại, in ra "false"

```python
def sum_divisors(n):
    s = 0
    for i in range(1, int(n**0.5) + 1):
        if n % i == 0:
            s += i
            if i*i != n and i != 1:
                s += n // i
    return s

a = int(input("Nhập số thứ nhất: "))
b = int(input("Nhập số thứ hai: "))

if sum_divisors(a) == b and sum_divisors(b) == a:
    print("true")
else:
    print("false")
```

W4A14 
[GCD] 
Cho hai số m và n là hai số nguyên dương nhập từ bàn phím. Viết chương trình 
tìm ước chung lớn nhất của hai số m và n.

```python
m = int(input("Nhập m: "))
n = int(input("Nhập n: "))

# Thuật toán Euclid sử dụng vòng lặp while
a, b = m, n
while b != 0:
    a, b = b, a % b

print(f"Ước chung lớn nhất của {m} và {n} là: {a}")
```

"Vừa gà vừa chó, 
Bó lại cho tròn, 
Ba mươi sáu con, 
Một trăm chân chẵn" 
Từ bài toán dân gian trên, mở rộng thành chương trình nhận đầu vào là tổng số 
con và tổng số chân của gà và chó. Nếu tìm được số phù hợp, in ra số lượng gà và 
số lượng chó. Ngược lại, in ra "invalid".

```python
tong_con = int(input("Nhập tổng số con: "))
tong_chan = int(input("Nhập tổng số chân: "))

found = False

# Duyệt qua tất cả khả năng của số gà (từ 0 đến tổng số con)
for ga in range(tong_con + 1):
    cho = tong_con - ga
    if (2 * ga + 4 * cho) == tong_chan:
        print(f"Số gà: {ga}, Số chó: {cho}")
        found = True
        break

if not found:
    print("invalid")
```
